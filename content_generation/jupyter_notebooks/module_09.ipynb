## 1. Instalando o Python

Para rodar um *script* fora do Colab — no seu próprio computador, ou em um servidor — você precisa ter o Python (e o **PIP**, seu gerenciador de pacotes) instalados na máquina que vai executar o código.

### 1.1 No Google Colab

O Python e o PIP **já vêm instalados** na máquina virtual que o Colab disponibiliza para você — não é preciso instalar nada.

**Exemplo:** conferindo a versão do Python a partir do próprio Python, com o módulo nativo `platform`.

````python
import platform

print(platform.python_version())
````

````python
print(platform.system())
````

> 💡 **Dica:** a máquina virtual do Colab roda **Linux**, então o resultado do código acima normalmente será `Linux` — mesmo que o seu computador seja Windows ou macOS. É um ambiente completamente à parte do seu computador.

Além de rodar código Python em uma célula, o Colab permite executar **comandos de terminal**: basta colocar um `!` antes do comando.

**Exemplo:** conferindo as versões do Python e do PIP diretamente pelo terminal do Colab.

````python
!python -V
````

````python
!pip -V
````

> 💡 O `!` não é Python: ele diz ao Colab "rode isto como um comando de terminal (Bash)", e não como código Python. Por isso essas duas últimas células não são executadas pelo verificador automático deste curso — elas só funcionam dentro de um notebook real, como o Colab.

### 1.2 Na sua máquina local

Antes de instalar, verifique se o Python já não está instalado. Abra o terminal (Windows: `cmd` ou `PowerShell`; macOS/Linux: `terminal`) e execute:

```shell
python -V
pip -V
```

> ⚠️ **Atenção:** no Windows, o comando costuma ser `python`. Em macOS e Linux, é comum ser `python3` (o nome `python` sozinho pode não existir, ou apontar para uma instalação antiga do Python 2). Se `python -V` não funcionar, tente `python3 -V`.

Se os comandos não forem reconhecidos, o Python não está instalado. Acesse o [site oficial do Python](https://www.python.org/), baixe o instalador para o seu sistema operacional e siga as instruções — este [tutorial do Real Python](https://realpython.com/installing-python/) é um ótimo guia passo a passo.

> ⚠️ **Atenção (Windows):** durante a instalação, marque a opção **"Add Python to PATH"**. Sem ela, o terminal não vai reconhecer o comando `python`, mesmo com o Python já instalado — um dos erros de configuração mais comuns entre iniciantes.

Depois de instalar, feche e reabra o terminal, e rode `python -V` novamente para confirmar.

## 2. Executando código no terminal

Além de rodar um arquivo de script inteiro, o Python tem um **modo interativo**, chamado **REPL** (*Read-Eval-Print Loop*): você digita uma linha de código, o Python executa e mostra o resultado na hora, uma linha de cada vez.

### 2.1 No Google Colab

Cada célula de código do Colab já funciona como um REPL: você escreve uma instrução, executa (`Shift + Enter`) e vê o resultado imediatamente, sem precisar rodar um arquivo `.py` inteiro. É por isso que o Colab é tão bom para aprender e testar código rapidamente.

### 2.2 Na sua máquina local

Para abrir o REPL do Python no seu computador:

1. Crie uma pasta para os arquivos do curso (sugestão);
2. Abra o terminal **dentro dessa pasta** (no Windows, `Shift + botão direito` na pasta; no macOS/Linux, botão direito e "Abrir terminal aqui");
3. Execute o comando:

```shell
python
```

O terminal muda para o *prompt* `>>>`, indicando que você está dentro do Python. A partir daí, qualquer linha digitada é executada na hora:

```python
>>> print("Olá, mundo!")
Olá, mundo!
>>> 2 + 2
4
```

Para sair do REPL, digite `quit()` (ou `exit()`) e pressione `Enter`, ou feche o terminal.

## 3. Criando arquivos de script

Um **arquivo de script** é um arquivo de texto com extensão `.py` contendo várias linhas de código Python, salvas juntas para serem executadas (e reexecutadas) de uma vez, sempre da mesma forma. É o formato usado para automações reais — algo que precisa rodar sozinho, sem um notebook aberto.

### 3.1 No Google Colab

O Colab não tem um editor de arquivos tradicional, mas você pode criar um arquivo `.py` diretamente de uma célula com o comando mágico `%%writefile nome_do_arquivo.py`: tudo o que vier **depois** dessa linha, na mesma célula, é salvo no arquivo.

**Exemplo:** um *script* que recebe um nome pela linha de comando e imprime uma saudação.

````python
%%writefile saudacao.py
import sys

nome = sys.argv[1]
print(f'Olá, {nome}!')
````

O módulo nativo `sys` guarda, na lista `sys.argv`, os **argumentos de linha de comando**: `sys.argv[0]` é sempre o nome do próprio script, e `sys.argv[1]`, `sys.argv[2]`... são os valores extras informados por quem executou o comando.

Agora executamos o arquivo recém-criado, passando `'Maria'` como argumento:

````python
!python saudacao.py Maria
````

> ⚠️ **Atenção:** se o script for executado sem o argumento esperado (`!python saudacao.py`, sem nome nenhum), `sys.argv` terá só um item (`sys.argv[0]`), e tentar acessar `sys.argv[1]` gera um `IndexError`. Scripts para uso real costumam validar se os argumentos necessários foram informados antes de usá-los.

**Exemplo (mundo real):** um *script* de automação que consulta uma taxa de juros em um site e registra o resultado, com data e hora, em um arquivo `.csv` — o tipo de tarefa que normalmente roda sozinha, em um horário programado (ex: todos os dias às 9h).

> 💡 Este é o **conteúdo** de um arquivo `extrair_taxa.py`, mostrado aqui só para leitura — ele não é executado por este notebook, pois depende de acesso à internet e de um serviço externo que pode mudar com o tempo. É exatamente esse tipo de tarefa, porém, que faz sentido rodar como *script*, e não dentro de um notebook.

```python
import os
import json
from datetime import datetime

import requests

URL = 'https://exemplo-api-financeira.com/taxa-cdi'
ARQUIVO_SAIDA = 'taxa-cdi.csv'

data_e_hora = datetime.now()
data = data_e_hora.strftime('%Y-%m-%d')
hora = data_e_hora.strftime('%H:%M:%S')

# Consultando a taxa no site
try:
    resposta = requests.get(URL)
    resposta.raise_for_status()
except requests.HTTPError:
    print('Taxa não encontrada, mas o script vai continuar.')
    taxa_cdi = None
except Exception as erro:
    print('Erro de conexão, interrompendo o script.')
    raise erro
else:
    dado = json.loads(resposta.text)
    taxa_cdi = float(dado['taxa'].replace(',', '.'))

# Criando o arquivo com cabeçalho, caso ainda não exista
if not os.path.exists(ARQUIVO_SAIDA):
    with open(file=ARQUIVO_SAIDA, mode='w', encoding='utf8') as arquivo:
        arquivo.write('data,hora,taxa\n')

# Só registra a linha se a taxa foi obtida com sucesso
if taxa_cdi is not None:
    with open(file=ARQUIVO_SAIDA, mode='a', encoding='utf8') as arquivo:
        arquivo.write(f'{data},{hora},{taxa_cdi}\n')
    print('Sucesso!')
```

> 💡 **Dica:** APIs de sites mudam com o tempo (endereço, formato da resposta etc.). Antes de usar uma API externa em um projeto real, sempre confira a documentação oficial e atualizada do serviço.

### 3.2 Na sua máquina local

A dinâmica é a mesma, só muda a ferramenta usada para criar o arquivo:

1. Crie o arquivo `.py` com um editor de texto ou de código (na próxima seção veremos as melhores opções);
2. Salve o arquivo na pasta onde você abriu o terminal;
3. Execute com o mesmo comando usado no Colab, sem o `!` (que só existe dentro do notebook):

```shell
python saudacao.py Maria
```

### 3.3 Ferramentas de desenvolvimento (IDEs)

Um bloco de notas simples funciona para escrever `.py`, mas uma **IDE** (*Integrated Development Environment*, Ambiente de Desenvolvimento Integrado) torna o trabalho muito mais produtivo: ela destaca erros de sintaxe antes de você executar o código, sugere autocompletar, permite depurar (*debug*) o código passo a passo e já vem com um terminal embutido.

As duas opções mais usadas no mercado são:

- **Visual Studio Code**, da Microsoft, gratuito e leve — a recomendação mais comum para começar ([link](https://code.visualstudio.com/));
- **PyCharm**, da JetBrains, mais robusto e focado especificamente em Python, com uma versão gratuita (*Community*) e uma paga (*Professional*) ([link](https://www.jetbrains.com/)).

> 💡 **Dica:** o próprio Visual Studio Code tem uma extensão oficial de Jupyter/Colab, que permite abrir e executar notebooks `.ipynb` (como os deste curso) sem sair do editor.
