## 1. Leitura de arquivos

Na vida real, os dados raramente são digitados à mão no código: eles chegam em **arquivos** exportados de sistemas, planilhas ou bancos de dados. Saber abrir e ler um arquivo é o primeiro passo de qualquer análise de dados.

### 1.1 Motivação

Você trabalha no time de dados de um banco e recebeu o arquivo `banco.csv`, com informações de alguns clientes. A primeira tarefa é simples: **descobrir a idade média dos clientes**.

Um arquivo **CSV** (*Comma-Separated Values*, ou "valores separados por vírgula") é um arquivo de texto em que cada linha representa um registro e as colunas são separadas por vírgulas. A primeira linha, chamada de **cabeçalho**, traz o nome das colunas. No nosso arquivo, elas estão em inglês:

| Coluna | Significado |
| --- | --- |
| `age` | Idade |
| `job` | Profissão |
| `marital` | Estado civil |
| `education` | Escolaridade |
| `default` | Está inadimplente? |
| `balance` | Saldo em conta |
| `housing` | Tem financiamento imobiliário? |
| `loan` | Tem empréstimo pessoal? |

**Como podemos levar o conteúdo desse arquivo para dentro do Python e calcular a idade média?**

### 1.2 Preparando o arquivo de exemplo

Antes de começar, precisamos que o arquivo `banco.csv` exista na pasta em que o notebook está rodando. A célula abaixo cria esse arquivo. Não se preocupe com os detalhes do código agora: ele usa exatamente o que vamos aprender na seção 2 (escrita de arquivos).

````python
# Conteúdo do arquivo: um cliente por linha, colunas separadas por vírgula
conteudo_banco = '''age,job,marital,education,default,balance,housing,loan
30,unemployed,married,primary,no,1787,no,no
33,services,married,secondary,no,4789,yes,yes
35,management,single,tertiary,no,1350,yes,no
30,management,married,tertiary,no,1476,yes,yes
59,blue-collar,married,secondary,no,0,yes,no
35,management,single,tertiary,no,747,no,no
36,self-employed,married,tertiary,no,307,yes,no
39,technician,married,secondary,no,147,yes,no
41,entrepreneur,married,tertiary,no,221,yes,no
43,services,married,primary,no,-88,yes,yes
'''

# Criando o arquivo banco.csv (veremos como isso funciona na seção 2)
with open('banco.csv', mode='w', encoding='utf-8') as arquivo:
    arquivo.write(conteudo_banco)

print('Arquivo banco.csv criado!')
````

> 💡 **Dica:** no Google Colab, clique no ícone de **pasta** (📁) na barra lateral esquerda para ver os arquivos da sessão. O `banco.csv` vai aparecer lá, e você pode até baixá-lo. Lembre-se de que esses arquivos são **temporários** e somem quando a sessão termina.

> 💡 **Dica:** você pode encontrar notebooks que criam arquivos com o comando `%%writefile nome_do_arquivo` no início da célula. Ele não é Python: é um "comando mágico" do Jupyter/Colab. Aqui usamos Python puro para que o código funcione em qualquer ambiente.

### 1.3 Abrindo arquivos com `with` e `open()`

Para trabalhar com um arquivo, primeiro precisamos **abri-lo**. Em Python, fazemos isso com a função nativa `open()`, quase sempre acompanhada da palavra-chave `with`:

```python
with open('caminho/do/arquivo', mode='modo', encoding='utf-8') as apelido:
    # bloco de código que usa o arquivo (indentado)
```

| Parte | O que significa |
| --- | --- |
| `'caminho/do/arquivo'` | Nome (ou caminho) do arquivo. Sem pasta, o Python procura na pasta atual. |
| `mode` | Modo de abertura. `'r'` (*read*) é leitura e é o padrão. Veremos os outros na seção 2. |
| `encoding` | Codificação do texto. Use `'utf-8'` para que acentos e cedilha funcionem corretamente. |
| `as apelido` | Nome da variável que representa o arquivo aberto dentro do bloco. |

O `with` garante que o arquivo seja **fechado automaticamente** ao final do bloco, mesmo que aconteça um erro no meio do caminho. Um arquivo esquecido aberto pode ficar bloqueado, consumir memória ou até perder dados que ainda não foram gravados.

> ⚠️ **Atenção:** se o arquivo não existir, `open()` no modo leitura gera um erro `FileNotFoundError`. Podemos tratar essa situação com o `try/except` que vimos no módulo anterior:

````python
# Tentando abrir um arquivo que não existe
try:
    with open('arquivo_que_nao_existe.csv', mode='r', encoding='utf-8') as arquivo:
        print(arquivo.read())
except FileNotFoundError:
    print('Arquivo não encontrado. Confira o nome e a pasta.')
````

### 1.4 Lendo tudo com `read()`

O método `.read()` lê **todo o conteúdo** do arquivo de uma vez e o devolve como uma única *string*:

````python
with open('banco.csv', mode='r', encoding='utf-8') as arquivo:
    conteudo = arquivo.read()

# Fora do bloco, o arquivo já foi fechado, mas o texto lido continua na variável
print(conteudo)
print(type(conteudo))
````

> 💡 **Dica:** no modo `'r'` o arquivo só pode ser lido. Qualquer tentativa de escrever nele gera um erro, o que protege o arquivo original contra alterações acidentais.

### 1.5 Lendo linha a linha

O `.read()` é prático, mas carrega o arquivo inteiro na memória. Para arquivos grandes (centenas de *megabytes* ou mais), o ideal é ler **uma linha por vez**. O Python oferece três formas:

| Forma | O que retorna | Quando usar |
| --- | --- | --- |
| `arquivo.readline()` | A **próxima linha** do arquivo, como *string* | Para ler uma linha específica, como o cabeçalho |
| `for linha in arquivo:` | Uma linha a cada repetição do laço | Para percorrer o arquivo inteiro sem carregá-lo todo na memória |
| `arquivo.readlines()` | Uma **lista** com todas as linhas | Arquivos pequenos, quando você precisa das linhas em uma lista |

**Exemplo:** o `.readline()` lê uma linha e "avança" no arquivo, então a chamada seguinte já devolve a próxima linha.

````python
with open('banco.csv', mode='r', encoding='utf-8') as arquivo:
    cabecalho = arquivo.readline()        # 1ª linha: o cabeçalho
    primeiro_cliente = arquivo.readline()  # 2ª linha: o primeiro cliente

print(cabecalho)
print(primeiro_cliente)
````

Repare que apareceu uma **linha em branco** depois de cada texto. Isso acontece porque cada linha lida termina com o caractere invisível `\n` (quebra de linha), e o `print()` acrescenta mais uma quebra. A função `repr()` mostra a *string* "crua", com esses caracteres visíveis:

````python
# repr() revela os caracteres invisíveis, como o \n no final da linha
print(repr(cabecalho))

# .strip() remove espaços e quebras de linha do início e do fim
print(repr(cabecalho.strip()))
````

**Exemplo:** percorrendo o arquivo inteiro com `for`, que é a forma mais simples e eficiente de ler linha a linha.

````python
with open('banco.csv', mode='r', encoding='utf-8') as arquivo:
    for linha in arquivo:
        # .strip() evita a linha em branco extra entre os registros
        print(linha.strip())
````

**Exemplo:** o `.readlines()` devolve todas as linhas em uma lista, cada uma ainda com o `\n` no final.

````python
with open('banco.csv', mode='r', encoding='utf-8') as arquivo:
    linhas = arquivo.readlines()

print(len(linhas))  # cabeçalho + 10 clientes
print(linhas[:3])   # as três primeiras linhas
````

> ⚠️ **Atenção:** quando o arquivo chega ao fim, o `.readline()` devolve uma *string* **vazia** (`''`), e não `None`. Você pode encontrar códigos que usam um laço `while` com `.readline()` para ler o arquivo, parando quando a *string* vem vazia. O `for linha in arquivo` faz exatamente o mesmo, com menos código e menos chance de erro.

### 1.6 Revisitando a motivação

Agora podemos calcular a idade média dos clientes. O plano é:

1. Ler e descartar o cabeçalho com `.readline()`;
2. Percorrer as demais linhas com `for`;
3. Em cada linha, remover o `\n` com `.strip()` e separar as colunas com `.split(',')`;
4. Pegar a primeira coluna (a idade), converter para `int` e guardar numa lista.

````python
idades = []

with open('banco.csv', mode='r', encoding='utf-8') as arquivo:
    arquivo.readline()  # descarta o cabeçalho

    for linha in arquivo:
        colunas = linha.strip().split(',')  # '30,unemployed,...' -> ['30', 'unemployed', ...]
        idade = int(colunas[0])             # a idade é a 1ª coluna e vem como texto
        idades.append(idade)

print(idades)
````

Com as idades em uma lista, a média é a soma dividida pela quantidade:

````python
idade_media = sum(idades) / len(idades)
print(f'Idade média dos clientes: {idade_media:.1f} anos')
````

> ⚠️ **Atenção:** tudo o que é lido de um arquivo de texto chega como `str`, mesmo que pareça número. Sem o `int()`, a lista teria `'30'`, `'33'` e assim por diante, e o `sum()` geraria um erro.

> 💡 **Dica:** separar colunas com `.split(',')` funciona bem para arquivos simples, mas falha quando um valor contém vírgula (por exemplo, `"São Paulo, SP"`). Para esses casos existem o módulo `csv` do Python e a biblioteca **Pandas**, que veremos em módulos futuros.

## 2. Escrita de arquivos

Tão importante quanto ler dados é **guardar resultados**: um relatório, uma lista filtrada, um arquivo para outro sistema. Nesta seção vamos criar arquivos, sobrescrevê-los e acrescentar conteúdo a eles.

### 2.1 Modos de abertura

O parâmetro `mode` do `open()` define o que podemos fazer com o arquivo:

| Modo | Nome | Se o arquivo existe... | Se o arquivo não existe... |
| --- | --- | --- | --- |
| `'r'` | Leitura (*read*) | Abre para leitura | Gera `FileNotFoundError` |
| `'w'` | Escrita (*write*) | **Apaga todo o conteúdo** e escreve do zero | Cria o arquivo |
| `'a'` | Acréscimo (*append*) | Escreve **no final**, mantendo o conteúdo | Cria o arquivo |
| `'x'` | Criação exclusiva | Gera `FileExistsError` | Cria o arquivo |

> ⚠️ **Atenção:** o modo `'w'` apaga o arquivo sem pedir confirmação. Se o objetivo é adicionar conteúdo, use `'a'`. Se quer ter certeza de não sobrescrever nada, use `'x'`.

### 2.2 Escrevendo com `write()`

O método `.write()` escreve uma *string* no arquivo. Ele **não** adiciona a quebra de linha automaticamente, então precisamos incluir o `\n` no final de cada linha.

**Exemplo:** salvando as idades extraídas na seção 1 em um novo arquivo, `idades.csv`.

````python
with open('idades.csv', mode='w', encoding='utf-8') as arquivo:
    arquivo.write('idade\n')  # cabeçalho

    for idade in idades:
        # write() só aceita str: convertemos o número e acrescentamos a quebra de linha
        arquivo.write(f'{idade}\n')

print('Arquivo idades.csv gravado!')
````

Vamos ler o arquivo de volta para conferir o que foi gravado:

````python
with open('idades.csv', mode='r', encoding='utf-8') as arquivo:
    print(arquivo.read())
````

> ⚠️ **Atenção:** passar um número direto para o `.write()`, como `arquivo.write(30)`, gera um `TypeError`. Converta com `str()` ou use uma *f-string*, como fizemos acima.

### 2.3 Acrescentando conteúdo com o modo `'a'`

**Exemplo:** dois novos clientes foram cadastrados e precisamos incluir suas idades no final do arquivo, sem perder as que já estão lá.

````python
novas_idades = [25, 48]

# Modo 'a': o conteúdo existente é mantido e o novo vai para o final
with open('idades.csv', mode='a', encoding='utf-8') as arquivo:
    for idade in novas_idades:
        arquivo.write(f'{idade}\n')

with open('idades.csv', mode='r', encoding='utf-8') as arquivo:
    print(arquivo.read())
````

> 💡 **Dica:** se você executar a célula acima mais de uma vez, as idades `25` e `48` serão acrescentadas de novo a cada execução. Para voltar ao estado inicial, execute novamente a célula da seção 2.2, que usa o modo `'w'` e recria o arquivo do zero.

### 2.4 Lendo de um arquivo e escrevendo em outro

É muito comum ler um arquivo, processar as linhas e gravar o resultado em outro. Um único `with` pode abrir os dois arquivos, separados por vírgula.

**Exemplo:** copiando o `banco.csv` para um arquivo com outra extensão, `banco.txt`.

````python
with open('banco.csv', mode='r', encoding='utf-8') as leitura, \
     open('banco.txt', mode='w', encoding='utf-8') as escrita:
    for linha in leitura:
        escrita.write(linha)  # a linha já termina com \n

print('Cópia concluída!')
````

> 💡 **Dica:** a extensão (`.csv`, `.txt`) é só parte do nome do arquivo e serve para indicar aos programas como abri-lo. O conteúdo dos dois arquivos é exatamente o mesmo texto.

**Exemplo:** o gerente de crédito pediu um arquivo só com os clientes que têm **empréstimo pessoal** (coluna `loan`, a 8ª coluna, com valor `yes`).

````python
with open('banco.csv', mode='r', encoding='utf-8') as leitura, \
     open('clientes_com_emprestimo.csv', mode='w', encoding='utf-8') as escrita:
    # O cabeçalho é copiado como está
    escrita.write(leitura.readline())

    for linha in leitura:
        colunas = linha.strip().split(',')
        tem_emprestimo = colunas[7] == 'yes'  # índice 7 = 8ª coluna (loan)
        if tem_emprestimo:
            escrita.write(linha)

with open('clientes_com_emprestimo.csv', mode='r', encoding='utf-8') as arquivo:
    print(arquivo.read())
````

## 3. Funções

Conforme os programas crescem, os mesmos trechos de código começam a se repetir. As **funções** permitem dar um nome a um bloco de código e executá-lo sempre que precisarmos, com valores diferentes a cada vez. Isso deixa o código mais curto, mais organizado e mais fácil de corrigir.

### 3.1 Motivação

Você trabalha em uma corretora de investimentos e precisa simular o retorno de aplicações com **juros compostos** em diversos cenários. No regime de juros compostos, a cada ano o rendimento é calculado sobre o valor acumulado até ali:

$$\text{valor no fim do ano} = \text{valor no início do ano} \times (1 + \text{taxa de juros anual})$$

Veja como fica o código para simular apenas **dois** cenários:

````python
# Cenário 1
valor_inicial = 1000.00
taxa_juros_anual = 0.05
anos = 10

valor_final = valor_inicial
for ano in range(anos):
    valor_final = valor_final * (1 + taxa_juros_anual)

print(f'R$ {valor_inicial:.2f} a {taxa_juros_anual:.0%} ao ano, em {anos} anos: R$ {valor_final:.2f}')

# Cenário 2: exatamente o mesmo código, só mudam os valores
valor_inicial = 1020.00
taxa_juros_anual = 0.03
anos = 10

valor_final = valor_inicial
for ano in range(anos):
    valor_final = valor_final * (1 + taxa_juros_anual)

print(f'R$ {valor_inicial:.2f} a {taxa_juros_anual:.0%} ao ano, em {anos} anos: R$ {valor_final:.2f}')
````

> 💡 **Dica:** na *f-string*, o formato `:.0%` multiplica o número por 100 e acrescenta o símbolo de porcentagem: `0.05` vira `5%`.

E se fossem 50 cenários? Copiar e colar o mesmo bloco 50 vezes deixaria o código enorme, e um erro no cálculo precisaria ser corrigido em 50 lugares.

**Como podemos reaproveitar esse código e evitar tanta repetição?**

### 3.2 Definição

Uma **função** é um bloco de código com nome, que **só é executado quando é chamado**. Já usamos várias funções prontas do Python, como `print()`, `len()` e `open()`. Agora vamos criar as nossas.

**Definindo** uma função:

```python
def nome_da_funcao(parametro_1, parametro_2):
    # bloco de código (indentado)
    return valor_de_retorno
```

**Chamando** uma função:

```python
resultado = nome_da_funcao(argumento_1, argumento_2)
```

| Parte | O que significa |
| --- | --- |
| `def` | Palavra-chave que inicia a definição de uma função |
| `nome_da_funcao` | Nome da função, em `snake_case` e de preferência com um verbo (`calcular_`, `ler_`, `exibir_`) |
| `(parametro_1, ...)` | **Parâmetros**: variáveis que recebem os valores enviados na chamada |
| `return` | Devolve um resultado para quem chamou a função |

**Exemplo:** uma função que exibe uma mensagem na tela.

````python
# Definir a função NÃO executa o código: apenas o guarda com um nome
def exibir_mensagem(mensagem):
    print(mensagem)
````

````python
# Agora sim, a função é executada (e pode ser chamada quantas vezes quisermos)
exibir_mensagem('Olá, pessoal! Bem-vindos ao módulo 04.')
exibir_mensagem('Funções evitam repetição de código.')
````

> ⚠️ **Atenção:** assim como as variáveis, a função precisa ser **definida antes** de ser chamada. Se você chamar `exibir_mensagem()` sem ter executado a célula com o `def`, o Python vai gerar um `NameError`.

> 💡 **Dica:** é uma boa prática descrever o que a função faz em uma ***docstring***, um texto entre aspas triplas logo abaixo do `def`. Ela aparece quando alguém usa `help()` na função.

````python
def exibir_mensagem(mensagem):
    '''Exibe a mensagem recebida na tela.'''
    print(mensagem)

help(exibir_mensagem)
````

### 3.3 Parâmetros e argumentos

**Parâmetros** são as variáveis declaradas entre parênteses no `def`. **Argumentos** são os valores que enviamos para esses parâmetros no momento da chamada. Uma função pode ter nenhum, um ou vários parâmetros.

**Exemplo:** função **sem parâmetros**.

````python
def obter_pi():
    return 3.14159265359

valor_pi = obter_pi()
print(valor_pi)
````

> ⚠️ **Atenção:** nunca guarde o resultado em uma variável com o **mesmo nome** da função, como `obter_pi = obter_pi()`. A variável substitui a função, e uma segunda chamada `obter_pi()` gera um `TypeError`, porque agora o nome aponta para um `float`, e não para uma função.

**Exemplo:** função **com vários parâmetros**. Os argumentos podem ser passados de duas formas:

- **Posicionais:** na mesma ordem dos parâmetros;
- **Nomeados:** informando `parametro=valor`. Nesse caso a ordem não importa, e a chamada fica mais legível.

````python
def calcular_total_compra(preco_unitario, quantidade):
    return preco_unitario * quantidade

# Argumentos posicionais: a ordem importa
print(calcular_total_compra(12.50, 4))

# Argumentos nomeados: a ordem não importa
print(calcular_total_compra(quantidade=4, preco_unitario=12.50))
````

**Exemplo:** parâmetros com **valor padrão**. Se o argumento não for informado na chamada, o parâmetro assume o valor definido no `def`.

````python
def calcular_preco_final(preco, desconto=0.0):
    return preco * (1 - desconto)

print(calcular_preco_final(100))                # usa o desconto padrão (0%)
print(calcular_preco_final(100, desconto=0.15))  # desconto de 15%
````

> ⚠️ **Atenção:** parâmetros com valor padrão devem vir **depois** dos parâmetros obrigatórios. Uma definição como `def calcular_preco_final(desconto=0.0, preco):` gera um `SyntaxError`.

Também podemos indicar o **tipo esperado** de cada parâmetro e do retorno, usando as ***type hints*** (dicas de tipo): `parametro: tipo` e `-> tipo_do_retorno`.

````python
def calcular_preco_final(preco: float, desconto: float = 0.0) -> float:
    return preco * (1 - desconto)

print(calcular_preco_final(250.0, desconto=0.10))
````

> 💡 **Dica:** as *type hints* servem como documentação e ajudam editores de código a apontar erros, mas o Python **não** as verifica durante a execução. Uma chamada com um tipo diferente do indicado continua funcionando (ou falhando) normalmente.

### 3.4 Retorno

O `return` devolve um valor para quem chamou a função. Esse valor pode ser guardado numa variável, impresso ou usado em outras contas.

````python
def converter_para_maiusculas(texto: str) -> str:
    texto_maiusculo = texto.upper()
    return texto_maiusculo

nome = 'André Perez'
nome_maiusculo = converter_para_maiusculas(nome)

print(nome)
print(nome_maiusculo)
````

**Exibir não é o mesmo que retornar.** Uma função sem `return` devolve automaticamente o valor `None`. Por isso, o resultado de uma função que apenas usa `print()` não pode ser aproveitado depois:

````python
def exibir_dobro(numero):
    print(numero * 2)   # apenas mostra na tela

def calcular_dobro(numero):
    return numero * 2   # devolve o valor

resultado_exibir = exibir_dobro(5)
resultado_calcular = calcular_dobro(5)

print(f'exibir_dobro devolveu:  {resultado_exibir}')
print(f'calcular_dobro devolveu: {resultado_calcular}')
````

O `return` também **encerra a função** imediatamente: nada do que estiver depois dele no mesmo caminho é executado. Isso permite retornar valores diferentes conforme uma condição.

**Exemplo:** classificando o saldo de um cliente do banco.

````python
def classificar_saldo(saldo: float) -> str:
    if saldo < 0:
        return 'negativo'   # a função termina aqui se o saldo for negativo
    if saldo == 0:
        return 'zerado'
    return 'positivo'

print(classificar_saldo(-88))
print(classificar_saldo(0))
print(classificar_saldo(1787))
````

Uma função pode **retornar mais de um valor**, separando-os por vírgula. Na verdade, o Python junta os valores em uma **tupla**, que podemos "desempacotar" em várias variáveis na chamada.

**Exemplo:** separando o usuário e o provedor de um e-mail.

````python
def extrair_usuario_e_provedor(email: str) -> tuple[str, str]:
    partes = email.split('@')
    usuario = partes[0]
    provedor = partes[1]
    return usuario, provedor

usuario, provedor = extrair_usuario_e_provedor('andre.perez@gmail.com')

print(usuario)
print(provedor)
````

### 3.5 Tratando erros dentro de funções

Funções que lidam com arquivos podem falhar por vários motivos: pasta inexistente, falta de permissão, dados em formato inesperado. Com `try/except`, a função pode tratar o erro e **informar** a quem chamou se a operação deu certo.

**Exemplo:** uma função reutilizável para gravar um arquivo CSV de uma coluna, que retorna `True` em caso de sucesso e `False` em caso de erro.

````python
def escrever_arquivo_csv(nome_arquivo: str, cabecalho: str, valores: list) -> bool:
    '''Grava um CSV de uma coluna. Retorna True se deu certo e False se houve erro.'''
    try:
        with open(nome_arquivo, mode='w', encoding='utf-8') as arquivo:
            arquivo.write(f'{cabecalho}\n')
            for valor in valores:
                arquivo.write(f'{valor}\n')
    except Exception as erro:
        print(f'Não foi possível gravar o arquivo: {erro}')
        return False

    return True
````

````python
# Chamada correta: valores é uma lista
gravou = escrever_arquivo_csv('idades_funcao.csv', 'idade', [30, 33, 35, 30, 59])
print(f'Gravou com sucesso? {gravou}')
````

````python
# Chamada com erro: 10 não é uma lista, então o for dentro da função falha
gravou = escrever_arquivo_csv('idades_funcao_erro.csv', 'idade', 10)
print(f'Gravou com sucesso? {gravou}')
````

> 💡 **Dica:** repare que o programa **não travou** no segundo caso. O erro foi tratado dentro da função, e o valor `False` permite que o restante do código decida o que fazer.

### 3.6 Revisitando a motivação

Agora podemos colocar o cálculo dos juros compostos em uma função. Repare que ela apenas **calcula e retorna** o valor. Exibir o resultado fica por conta de quem a chama, o que deixa a função mais fácil de reaproveitar (por exemplo, para gravar os resultados em um arquivo em vez de imprimi-los).

````python
def calcular_juros_compostos(valor_inicial: float, taxa_juros_anual: float, anos: int) -> float:
    '''Retorna o valor final de um investimento com juros compostos anuais.'''
    valor_final = valor_inicial
    for _ in range(anos):  # "_" indica que a variável do laço não é usada
        valor_final = valor_final * (1 + taxa_juros_anual)
    return valor_final
````

Os mesmos dois cenários de antes, agora com uma linha cada:

````python
valor_final_1 = calcular_juros_compostos(valor_inicial=1000.00, taxa_juros_anual=0.05, anos=10)
valor_final_2 = calcular_juros_compostos(valor_inicial=1020.00, taxa_juros_anual=0.03, anos=10)

print(f'Cenário 1: R$ {valor_final_1:.2f}')
print(f'Cenário 2: R$ {valor_final_2:.2f}')
````

E simular vários cenários passa a ser trivial. **Exemplo:** comparando diferentes taxas para o mesmo valor inicial.

````python
taxas_simuladas = [0.03, 0.05, 0.08, 0.10]

for taxa in taxas_simuladas:
    valor_final = calcular_juros_compostos(valor_inicial=1000.00, taxa_juros_anual=taxa, anos=10)
    print(f'Taxa de {taxa:.0%} ao ano: R$ {valor_final:.2f}')
````

> 💡 **Dica:** os juros compostos também têm uma fórmula direta, $\text{valor final} = \text{valor inicial} \times (1 + \text{taxa})^{\text{anos}}$, que em Python fica `valor_inicial * (1 + taxa_juros_anual) ** anos`. Como o cálculo está isolado em uma função, poderíamos trocar o laço pela fórmula **sem mudar nenhuma chamada**. Essa é outra grande vantagem das funções.

````python
# Conferindo: a fórmula direta dá o mesmo resultado que o laço
print(round(1000.00 * (1 + 0.05) ** 10, 2))
print(round(calcular_juros_compostos(1000.00, 0.05, 10), 2))
````

> ⚠️ **Atenção:** o `round()` do Python usa o **arredondamento bancário**: quando o número está exatamente no meio (`.5`), ele arredonda para o número **par** mais próximo. Por isso `round(2.5)` resulta em `2`, e não `3`. Na maioria dos casos isso não faz diferença, mas é bom saber para não se surpreender.

````python
print(round(10 / 4))  # 2.5 -> 2 (o par mais próximo)
print(round(3.5))     # 3.5 -> 4 (o par mais próximo)
print(round(2.6))     # 2.6 -> 3 (não está no meio: arredonda normalmente)
````

## 4. Escopo

Com funções, surge uma pergunta importante: uma variável criada dentro de uma função existe fora dela? E o contrário? O **escopo** define **onde** uma variável pode ser acessada e **por quanto tempo** ela existe. Entender isso evita erros difíceis de encontrar.

### 4.1 Definição

O **escopo** de uma variável é a região do código em que ela é visível. Em Python, os dois escopos mais importantes são:

| Escopo | Onde a variável é criada | Onde pode ser usada | Quanto tempo existe |
| --- | --- | --- | --- |
| **Local** | Dentro de uma função (incluindo os parâmetros) | Apenas dentro daquela função | Enquanto a função está sendo executada |
| **Global** | Fora de qualquer função (no "corpo" do notebook) | Em qualquer lugar, inclusive dentro das funções (para leitura) | Até o programa (ou a sessão) terminar |

### 4.2 Escopo local

Variáveis criadas dentro de uma função são **locais**: elas nascem quando a função é chamada e são descartadas quando ela termina.

**Exemplo:** somando os valores de uma lista.

````python
def somar_lista(numeros: list) -> int:
    total = 0  # variável local: só existe dentro da função
    for numero in numeros:
        total = total + numero
    return total

soma = somar_lista([2, 4, 6, 8])
print(soma)
````

Fora da função, a variável `total` **não existe**. Tentar usá-la gera um erro:

````python
# print(total)    # ❌ NameError: name 'total' is not defined

# ✅ O resultado chega aqui fora pelo return, guardado na variável soma
print(soma)
````

> 💡 **Dica:** o Python já tem a função nativa `sum()`, que faz exatamente o que a `somar_lista()` faz. Criamos a nossa apenas para entender o escopo, mas no dia a dia prefira as funções nativas.

### 4.3 Escopo global

Variáveis criadas fora das funções são **globais** e podem ser **lidas** dentro de qualquer função:

````python
taxa_administracao = 0.02  # variável global

def calcular_valor_liquido(valor_bruto: float) -> float:
    # A função consegue LER a variável global
    return valor_bruto * (1 - taxa_administracao)

print(calcular_valor_liquido(1000))
````

Mas atenção: se a função **atribuir** um valor a uma variável com o mesmo nome, o Python cria uma **nova variável local**, e a global continua intacta:

````python
def simular_taxa_maior(valor_bruto: float) -> float:
    taxa_administracao = 0.10  # cria uma variável LOCAL com o mesmo nome
    return valor_bruto * (1 - taxa_administracao)

print(simular_taxa_maior(1000))  # usa a taxa local (10%)
print(taxa_administracao)        # a global continua 0.02
````

Para **alterar** uma variável global dentro de uma função, é preciso declará-la com a palavra-chave `global`:

````python
contador_simulacoes = 0

def registrar_simulacao():
    # contador_simulacoes += 1  # ❌ sem "global": UnboundLocalError
    global contador_simulacoes
    contador_simulacoes += 1

registrar_simulacao()
registrar_simulacao()
print(contador_simulacoes)
````

> ⚠️ **Atenção:** usar `global` torna o código difícil de entender e testar, porque a função passa a depender de algo "escondido" fora dela. Prefira o caminho mais claro: **receber os dados por parâmetros e devolver o resultado com `return`**.

````python
# ✅ Mesmo resultado, sem global: a função recebe o valor e devolve o novo
def incrementar(contador: int) -> int:
    return contador + 1

contador = 0
contador = incrementar(contador)
contador = incrementar(contador)
print(contador)
````

### 4.4 Estruturas condicionais e de repetição

Diferente de outras linguagens, em Python o `if` e o `for` **não criam um escopo próprio**. Uma variável criada dentro deles continua existindo depois que o bloco termina.

````python
saldo = 1787

if saldo > 0:
    situacao = 'positivo'  # criada dentro do if...
else:
    situacao = 'negativo'

print(situacao)  # ...mas acessível fora dele
````

````python
for indice in range(3):
    ultimo_processado = indice

# Depois do laço, as duas variáveis continuam existindo com o último valor
print(indice)
print(ultimo_processado)
````

> ⚠️ **Atenção:** uma variável criada em um bloco que **não foi executado** simplesmente não existe. Se no exemplo acima só o `if` definisse `situacao` (sem o `else`) e o saldo fosse negativo, o `print(situacao)` geraria um `NameError`. Por isso, garanta que a variável receba um valor em todos os caminhos possíveis, ou crie-a com um valor inicial antes do `if`.

````python
saldo = -88

# ✅ Valor inicial definido antes do if: a variável sempre existe
situacao = 'não negativo'
if saldo < 0:
    situacao = 'negativo'

print(situacao)
````

## Resumo do Módulo

| Conceito | Para que serve | Exemplo |
| --- | --- | --- |
| `with open(...) as arquivo:` | Abre um arquivo e o fecha automaticamente ao fim do bloco | `with open('banco.csv', mode='r', encoding='utf-8') as arquivo:` |
| `.read()` | Lê o arquivo inteiro como uma *string* | `conteudo = arquivo.read()` |
| `.readline()` | Lê a próxima linha | `cabecalho = arquivo.readline()` |
| `for linha in arquivo:` | Percorre o arquivo linha a linha | `for linha in arquivo: ...` |
| `.readlines()` | Lê todas as linhas em uma lista | `linhas = arquivo.readlines()` |
| `.write()` | Escreve uma *string* (sem quebra de linha automática) | `arquivo.write(f'{idade}\n')` |
| Modos `'r'`, `'w'`, `'a'`, `'x'` | Leitura, escrita (apaga), acréscimo e criação exclusiva | `open('idades.csv', mode='a')` |
| `def` | Define uma função | `def calcular_dobro(numero):` |
| Parâmetro com valor padrão | Torna o argumento opcional | `def calcular_preco_final(preco, desconto=0.0):` |
| `return` | Devolve um resultado e encerra a função | `return numero * 2` |
| Escopo local | Variável criada na função só existe dentro dela | `total` em `somar_lista()` |
| Escopo global | Variável criada fora das funções pode ser lida em qualquer lugar | `taxa_administracao` |

Funções e métodos vistos neste módulo: `open()`, `.read()`, `.readline()`, `.readlines()`, `.write()`, `.strip()`, `.split()`, `repr()`, `help()`, `sum()` e `round()`.
