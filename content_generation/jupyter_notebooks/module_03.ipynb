## 1. Estrutura condicional if / elif / else

A maioria dos programas precisa se comportar de forma diferente dependendo da situação: aprovar ou recusar um pagamento, liberar ou bloquear um acesso, aplicar um desconto ou não. Esse comportamento é controlado pelo **fluxo condicional**, e em Python ele é escrito com as palavras-chave `if`, `elif` e `else`.

### 1.1 Motivação

Você está desenvolvendo o sistema de pagamento com cartão de crédito de um *e-commerce*. Antes de aprovar a compra, o sistema precisa comparar o código de segurança (CVV) informado pelo cliente com o código cadastrado no banco.

````python
# Código digitado pelo cliente e código salvo no cadastro
codigo_de_seguranca_informado = '291'
codigo_de_seguranca_cadastro = '010'
````

**Como fazemos o programa executar um trecho de código diferente dependendo se os dois códigos são iguais ou não?**

### 1.2 if / else

A estrutura `if` / `else` executa um bloco de código **se** uma condição for `True`, e outro bloco **senão** (quando ela for `False`):

```python
if condicao:
  # executado quando condicao é True
else:
  # executado quando condicao é False
```

> ⚠️ **Atenção:** o Python usa a **indentação** (o recuo no início da linha) para saber o que pertence a cada bloco. Todas as linhas de um mesmo bloco precisam ter o mesmo recuo, ou o código não roda.

````python
esta_chovendo = False

if esta_chovendo:
  print('Leve um guarda-chuva')
else:
  print('Não precisa de guarda-chuva')
````

Aplicando ao nosso problema: comparamos os dois códigos com `==` (visto no Módulo 01) e usamos o resultado como condição.

````python
if codigo_de_seguranca_informado == codigo_de_seguranca_cadastro:
  print('Pagamento aprovado')
else:
  print('Erro: código de segurança inválido')
````

### 1.3 Condições compostas com and / or

Muitas vezes é preciso avaliar **mais de uma condição** ao mesmo tempo. Para isso, combinamos os operadores lógicos `and` e `or` (vistos no Módulo 01) diretamente na condição do `if`.

**Exemplo:** além do código de segurança, o sistema também exige que a senha informada esteja correta. O pagamento só é aprovado se **ambas** as condições forem verdadeiras.

````python
codigo_de_seguranca_informado = '852'
codigo_de_seguranca_cadastro = '852'

senha_informada = '7783'
senha_cadastro = '7783'
````

````python
# Os parênteses em cada comparação deixam a condição mais fácil de ler
if (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Pagamento aprovado')
else:
  print('Erro: pagamento não autorizado')
````

> 💡 Você também pode encontrar os operadores `&` (e) e `|` (ou) em bibliotecas como o **Pandas**. No Python "puro", porém, dentro de um `if`, a forma recomendada e mais legível é usar `and` e `or`.

### 1.4 if / elif / else

Quando existem **mais de duas** possibilidades, usamos `elif` (abreviação de "*else if*") para encadear várias condições. O Python testa cada condição **na ordem em que aparece** e executa o bloco da **primeira** que for verdadeira — ignorando todas as demais, mesmo que também sejam verdadeiras.

```python
if condicao_1:
  # executado se condicao_1 for True
elif condicao_2:
  # executado se condicao_1 for False e condicao_2 for True
else:
  # executado se nenhuma das anteriores for True
```

### 1.5 Revisitando a motivação

Agora temos tudo para tratar as **quatro** combinações possíveis entre código de segurança e senha, cada uma com uma mensagem diferente:

| Código correto? | Senha correta? | Resultado |
| --- | --- | --- |
| Sim | Sim | Pagamento aprovado |
| Não | Sim | Erro: código de segurança inválido |
| Sim | Não | Erro: senha inválida |
| Não | Não | Erro: código de segurança e senha inválidos |

````python
codigo_de_seguranca_informado = '802'
codigo_de_seguranca_cadastro = '852'

senha_informada = '7703'
senha_cadastro = '7783'
````

````python
if (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Pagamento aprovado')
elif (codigo_de_seguranca_informado != codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Erro: código de segurança inválido')
elif (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada != senha_cadastro):
  print('Erro: senha inválida')
else:
  print('Erro: código de segurança e senha inválidos')
````

> 💡 Experimente alterar os valores de `codigo_de_seguranca_informado` e `senha_informada` acima e executar as duas células novamente, para ver as outras três mensagens aparecerem.

## 2. Tratamento de exceções com try / except / finally

Mesmo um código sem erros de digitação pode falhar **durante a execução**, por causa de situações inesperadas: uma divisão por zero, um dado no formato errado, uma posição que não existe em uma lista. Esses erros de execução são chamados de **exceções**, e o Python permite tratá-las para que o programa não seja interrompido.

### 2.1 Motivação

Você está calculando quanto cada pessoa deve pagar em uma conta de restaurante, dividindo o valor total pela quantidade de pessoas. Só que esse número de pessoas vem de um formulário preenchido por outra pessoa, e às vezes chega errado — inclusive como zero.

Se executarmos a divisão diretamente, o Python interrompe o programa:

```python
preco_total_conta = 132.85
quantidade_de_pessoas = 0

valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
```

```
ZeroDivisionError: float division by zero
```

**Como evitamos que o programa trave quando isso acontecer?**

### 2.2 Tipos comuns de exceção

Cada tipo de erro tem um nome próprio, o que ajuda a tratá-lo de forma específica:

| Exceção | Ocorre quando... | Exemplo |
| --- | --- | --- |
| `ZeroDivisionError` | uma divisão tem divisor igual a zero | `10 / 0` |
| `TypeError` | uma operação combina tipos incompatíveis | `'Idade: ' + 19` |
| `ValueError` | um valor não pode ser convertido para o tipo pedido | `int('abc')` |
| `IndexError` | uma posição não existe em uma lista | `lista[10]` (lista com 3 itens) |
| `KeyError` | uma chave não existe em um dicionário | `dicionario['chave_inexistente']` |
| `NameError` | uma variável usada não foi definida | `print(variavel_que_nao_existe)` |

### 2.3 try / except

A estrutura `try` / `except` executa um bloco de código e, **se** uma exceção ocorrer, executa outro bloco no lugar — sem interromper o programa.

```python
try:
  # código que pode falhar
except TipoDoErro:
  # executado somente se essa exceção ocorrer
```

````python
preco_total_conta = 132.85
quantidade_de_pessoas = 0

try:
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Cada pessoa paga R$ {valor_por_pessoa:.2f}')
except ZeroDivisionError:
  print('Erro: a quantidade de pessoas precisa ser maior que zero.')
````

Com uma quantidade de pessoas válida, o `try` roda normalmente e o `except` nem é avaliado:

````python
preco_total_conta = 132.85
quantidade_de_pessoas = 2

try:
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Cada pessoa paga R$ {valor_por_pessoa:.2f}')
except ZeroDivisionError:
  print('Erro: a quantidade de pessoas precisa ser maior que zero.')
````

### 2.4 Capturando detalhes da exceção e tratando vários tipos

Podemos capturar a exceção em uma variável (convenção: `except ... as exc`) para inspecionar sua mensagem, e também tratar **vários tipos de exceção diferentes** no mesmo `try`, cada um com seu próprio `except`.

**Exemplo:** buscando uma posição que não existe em uma lista de anos.

````python
anos_com_vendas = [2019, 2020, 2021]

try:
  ano_mais_recente = anos_com_vendas[5]
  print(ano_mais_recente)
except IndexError as exc:
  print(f'Erro: posição inválida. A lista tem apenas {len(anos_com_vendas)} anos.')
  print(f'Mensagem original do Python: {exc}')
````

**Exemplo:** convertendo um texto vindo de um formulário para número, tratando o erro específico primeiro e um erro genérico por último.

> ⚠️ **Atenção:** a ordem importa. O Python testa os blocos `except` na ordem em que aparecem, então coloque sempre os erros **mais específicos primeiro** e um `except Exception` **por último**, como uma rede de segurança.

````python
dado_de_entrada = 'não é um número'

try:
  valor_convertido = int(dado_de_entrada)
  print(valor_convertido)
except ValueError:
  print('Erro: não foi possível converter o texto para número.')
except Exception as exc:
  print(f'Erro inesperado: {exc}')
````

### 2.5 try / except / finally

O bloco `finally` é executado **sempre**, tenha ocorrido uma exceção ou não. É útil para ações que precisam acontecer independentemente do resultado, como confirmar que um processo terminou.

````python
nome_cliente = 'André'
idade_cliente = 19

try:
  # idade_cliente é int; o + não junta str com int (TypeError)
  apresentacao = 'Cliente: ' + nome_cliente + ', idade: ' + idade_cliente
except TypeError:
  apresentacao = 'Cliente: ' + nome_cliente + ', idade: ' + str(idade_cliente)
finally:
  print(apresentacao)
  print('Cadastro processado.')
````

### 2.6 Revisitando a motivação

Vamos deixar o cálculo do valor por pessoa mais realista: a quantidade de pessoas chega como **texto** do formulário, então pode falhar de duas formas diferentes — um texto que não é número (`ValueError`) ou o número zero (`ZeroDivisionError`). Tratamos as duas, e usamos o `finally` para avisar que o cálculo terminou.

````python
preco_total_conta = 132.85
quantidade_de_pessoas_texto = '0'  # veio como texto de um formulário

try:
  quantidade_de_pessoas = int(quantidade_de_pessoas_texto)
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Cada pessoa paga R$ {valor_por_pessoa:.2f}')
except ValueError:
  print('Erro: a quantidade de pessoas precisa ser um número.')
except ZeroDivisionError:
  print('Erro: a quantidade de pessoas precisa ser maior que zero.')
finally:
  print('Cálculo finalizado.')
````

## 3. Estrutura de repetição for

Boa parte do trabalho de um programa é repetitivo: somar todos os valores de uma lista, verificar cada linha de uma planilha, processar cada cliente de um cadastro. Em vez de repetir o código manualmente, usamos a estrutura de repetição `for`.

### 3.1 Motivação

Você tem as vendas do dia de uma loja e quer saber duas coisas: o **total vendido** e **quantas vendas** ultrapassaram R\$ 300,00.

````python
vendas_do_dia = [361, 553, 194, 13, 510, 33, 135]
````

**Como processamos todos esses valores sem escrever uma linha de código para cada um?**

### 3.2 for / range

A estrutura `for` executa um bloco de código **uma vez para cada elemento** de uma coleção. Combinada com a função nativa `range()`, ela repete o bloco um número determinado de vezes:

```python
for variavel_temporaria in range(inicio, fim, passo):
  # executado uma vez para cada valor no intervalo
```

````python
for valor in range(6):
  print(valor)
````

> 💡 `range(fim)` começa em 0; `range(inicio, fim)` define o início; `range(inicio, fim, passo)` também define o incremento. Assim como no fatiamento de *strings* (Módulo 01), o valor `fim` **nunca** é incluído.

````python
# Múltiplos de 3, de 0 (incluído) até 15 (excluído)
for multiplo_de_tres in range(0, 15, 3):
  print(multiplo_de_tres)
````

### 3.3 for / listas

A forma mais comum de usar o `for` é para percorrer os elementos de uma lista, um a um:

````python
for venda in vendas_do_dia:
  print(venda)
````

Um padrão muito comum é **acumular** um resultado a cada repetição, usando uma variável criada **antes** do `for`:

````python
total_vendido = 0

for venda in vendas_do_dia:
  total_vendido += venda

print(f'Total vendido: R$ {total_vendido:.2f}')
````

E podemos combinar o `for` com o `if` (visto na seção 1) para contar apenas os elementos que atendem a uma condição:

````python
quantidade_vendas_acima_300 = 0

for venda in vendas_do_dia:
  if venda > 300:
    quantidade_vendas_acima_300 += 1

print(f'Vendas acima de R$ 300: {quantidade_vendas_acima_300}')
````

> 💡 *Strings* também são sequências e podem ser percorridas caractere a caractere. O operador `in` aqui testa se um caractere **pertence** a outro texto.

````python
frase = 'Ready To Deploy'
quantidade_vogais = 0

for caractere in frase.lower():
  if caractere in 'aeiou':
    quantidade_vogais += 1

print(f'Quantidade de vogais: {quantidade_vogais}')
````

### 3.4 for / dicionários

Dicionários possuem três métodos para serem percorridos em um `for`: `.items()` (chave e valor), `.keys()` (só as chaves) e `.values()` (só os valores).

````python
escore_de_credito = {'123.456.789-00': 750, '987.654.321-00': 812, '111.222.333-44': 980}
````

````python
# .items(): retorna cada par chave-valor
for documento, escore in escore_de_credito.items():
  print(f'Documento {documento}: escore {escore}')
````

````python
# .keys(): retorna apenas as chaves
for documento in escore_de_credito.keys():
  print(documento)
````

````python
# .values(): retorna apenas os valores
for escore in escore_de_credito.values():
  print(escore)
````

> 💡 `for documento in escore_de_credito:` (sem `.keys()`) tem exatamente o mesmo efeito, pois percorrer um dicionário diretamente já retorna suas chaves.

### 3.5 break / continue

Duas palavras-chave controlam o andamento de um laço `for`:

| Palavra-chave | Efeito |
| --- | --- |
| `break` | interrompe o laço **imediatamente**, sem executar as repetições restantes |
| `continue` | pula direto para a **próxima repetição**, sem executar o restante do bloco atual |

**Exemplo:** parar assim que encontrarmos a primeira venda acima de R\$ 500,00.

````python
for venda in vendas_do_dia:
  if venda > 500:
    print(f'Primeira venda acima de R$ 500: {venda}')
    break
````

**Exemplo:** exibir apenas as vendas com valor **par**, pulando as ímpares com `continue`.

````python
for venda in vendas_do_dia:
  if venda % 2 != 0:
    continue
  print(f'Venda par: {venda}')
````

> ⚠️ **Atenção:** qualquer código escrito **depois** de um `break` ou `continue`, ainda dentro do mesmo bloco `if`, nunca é executado — o laço já foi interrompido ou avançado antes de chegar lá.

### 3.6 Revisitando a motivação

Com `for`, `if` e o acúmulo em variáveis, resolvemos as duas perguntas da motivação em um único laço:

````python
vendas_do_dia = [361, 553, 194, 13, 510, 33, 135]

total_vendido = 0
quantidade_vendas_acima_300 = 0

for venda in vendas_do_dia:
  total_vendido += venda
  if venda > 300:
    quantidade_vendas_acima_300 += 1

print(f'Total vendido: R$ {total_vendido:.2f}')
print(f'Vendas acima de R$ 300: {quantidade_vendas_acima_300}')
````

## Resumo do Módulo

| --- | --- | --- |
| Condicional | Executar um bloco ou outro dependendo de uma condição | `if`, `elif`, `else` |
