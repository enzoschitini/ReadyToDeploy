## 1. Listas

Quando precisamos guardar **várias informações do mesmo tipo de evento** — como as transações de um dia — criar uma variável para cada uma não escala. As **listas** resolvem exatamente esse problema.

### 1.1 Motivação

O aplicativo do seu banco registra toda a sua movimentação financeira. No final do dia, o app soma todas as transações ao saldo inicial para calcular o **saldo final**. Guardando cada transação em uma variável separada, o código fica assim:

```python
saldo_inicial_dia_11 = 1000

transacao_1 = 243
transacao_2 = -798.58
transacao_3 = 427.12
transacao_4 = -10.91
```

```python
saldo_final_dia_11 = saldo_inicial_dia_11 + transacao_1 + transacao_2 + transacao_3 + transacao_4
print(round(saldo_final_dia_11, 2))
```

Funciona, mas repare no problema: se o dia tivesse 50 transações, seriam 50 variáveis diferentes, e o cálculo do saldo teria 50 termos escritos à mão. **Existe uma forma melhor de armazenar essas transações?**

### 1.2 Definição

**Listas** armazenam sequências **ordenadas** e **mutáveis** (podem ser alteradas depois de criadas) de valores, que podem se repetir. São do tipo `list` e são criadas entre colchetes `[]`, com os itens separados por vírgula.

```python
dados_usuario_web = ['André Perez', 'andre.perez', 'andre123', 'andre.perez@gmail.com']

print(dados_usuario_web)
print(type(dados_usuario_web))
```

Uma lista pode misturar valores de **tipos diferentes** — números, textos, booleanos e até outras listas:

```python
idade = 20
saldo_em_conta = 723.15
usuario_logado = True

dados_usuario_web = ['André Perez', idade, 'andre.perez', 'andre.perez@gmail.com', saldo_em_conta, usuario_logado]

print(dados_usuario_web)
print(type(dados_usuario_web))
```

### 1.3 Operações

**Concatenação (`+`)**: junta duas listas em uma só, mantendo a ordem.

**Exemplo:** unindo fabricantes de *hardware* mobile por região.

```python
fabricantes_china = ['xiaomi', 'huawei']
fabricantes_eua = ['apple', 'motorola']

fabricantes_mobile = fabricantes_china + fabricantes_eua
print(fabricantes_mobile)
```

**Fatiamento (*slicing*)**: assim como nas *strings*, acessamos itens de uma lista pela posição (índice), que começa em **0**. Índices negativos contam a partir do final.

```python
print(f'Posição 0:  {fabricantes_mobile[0]}')
print(f'Posição -1: {fabricantes_mobile[-1]}')
```

E também podemos extrair um **intervalo** de itens com `lista[inicio:fim]` — o item da posição `fim` não é incluído:

```python
fabricantes_china = fabricantes_mobile[0:2]
fabricantes_eua = fabricantes_mobile[2:len(fabricantes_mobile)]

print(f'China: {fabricantes_china}')
print(f'EUA:   {fabricantes_eua}')
```

Diferente das *strings*, as listas são **mutáveis**: podemos alterar um item existente atribuindo um novo valor à sua posição.

```python
print(fabricantes_mobile)

fabricantes_mobile[2] = 'nokia'
print(fabricantes_mobile)
```

### 1.4 Métodos

Os métodos abaixo são os mais usados no dia a dia para adicionar e remover itens de uma lista:

| Método | O que faz |
| --- | --- |
| `.append(valor)` | Adiciona um item ao **final** da lista |
| `.insert(indice, valor)` | Insere um item em uma **posição específica** |
| `.remove(valor)` | Remove a **primeira ocorrência** de um valor |
| `.pop(indice)` | Remove o item de uma posição e **retorna** o valor removido |

```python
juros_mensais = [0.05, 0.07, 0.02, 0.04, 0.08]
print(juros_mensais)
```

```python
# .append(): adiciona ao final
juros_mensais.append(0.09)
print(juros_mensais)
```

```python
# .insert(): adiciona em uma posição específica, empurrando os demais itens
juros_mensais.insert(0, 0.10)
print(juros_mensais)
```

```python
# .remove(): remove pelo VALOR (a primeira ocorrência)
juros_mensais.remove(0.10)
print(juros_mensais)
```

```python
# .pop(): remove pelo ÍNDICE e devolve o valor removido
terceiro_juros = juros_mensais.pop(2)
print(f'Valor removido: {terceiro_juros}')
print(juros_mensais)
```

> 💡 **Dica:** `.pop()` sem argumento remove e retorna o **último** item da lista — útil para simular uma pilha de itens.

### 1.5 Conversão

Alguns tipos podem ser convertidos em lista com a função `list()`. Uma *string*, por exemplo, vira uma lista com um item para cada caractere:

```python
email = 'andre.perez@gmail.com'
caracteres_email = list(email)

print(email)
print(caracteres_email)
```

### 1.6 Revisitando a motivação

Agora podemos guardar as transações do dia em uma **única lista**, e usar `.append()` para adicionar cada nova transação:

```python
transacoes_dia_11 = []

transacoes_dia_11.append(243)
transacoes_dia_11.append(-798.58)
transacoes_dia_11.append(427.12)
transacoes_dia_11.append(-10.91)

print(transacoes_dia_11)
```

E a função nativa `sum()` soma todos os itens de uma lista numérica de uma vez, sem precisar somar item por item:

```python
saldo_final_dia_11 = saldo_inicial_dia_11 + sum(transacoes_dia_11)
print(round(saldo_final_dia_11, 2))
```

> 💡 Não importa se o dia tem 4 ou 400 transações: o código continua exatamente o mesmo, pois `sum()` soma todos os itens da lista automaticamente.

## 2. Conjuntos

Listas guardam itens repetidos e na ordem em que foram inseridos. Mas, às vezes, o que importa não é a ordem nem a repetição — é só saber **quais valores diferentes** existem. Para isso usamos os **conjuntos**.

### 2.1 Motivação

Você trabalha como analista de mídias sociais e precisa descobrir **todas as hashtags diferentes** que entraram nos *trending topics* de uma rede social durante a semana. Você já tem as hashtags de cada dia, em listas separadas:

```python
hashtags_segunda = ['#tiago', '#joao', '#bbb']
hashtags_terca = ['#sarah', '#bbb', '#fiuk']
hashtags_quarta = ['#gil', '#thelma', '#lourdes']
hashtags_quinta = ['#rafa', '#fora', '#danilo']
hashtags_sexta = ['#juliete', '#arthur', '#bbb']
```

Se simplesmente concatenarmos as listas da semana, a hashtag `#bbb` (que apareceu em três dias diferentes) vai se repetir no resultado:

```python
hashtags_semana = hashtags_segunda + hashtags_terca + hashtags_quarta + hashtags_quinta + hashtags_sexta

print(hashtags_semana)
print(f'Total: {len(hashtags_semana)}')
```

Repare que o total (15) não representa a quantidade de hashtags **diferentes** — `#bbb` foi contada três vezes. **Como eliminamos as repetições?**

### 2.2 Definição

**Conjuntos** armazenam sequências **desordenadas** de valores, **sem repetição**. São do tipo `set` e são criados entre chaves `{}`.

```python
frutas = {'banana', 'maca', 'uva', 'uva'}

print(frutas)
print(type(frutas))
```

Repare que `'uva'` foi escrita duas vezes, mas aparece **uma única vez** no conjunto. Como conjuntos não têm ordem, a sequência exibida pode variar a cada execução.

> ⚠️ **Atenção:** por não terem ordem, conjuntos **não podem ser acessados por índice** (`frutas[0]` gera um erro `TypeError`).

### 2.3 Operações

**Diferença (`-`)**: retorna os itens que estão no primeiro conjunto, mas **não** estão no segundo.

**Exemplo:** países do norte da Europa que não são escandinavos.

```python
norte_europa = {'reino unido', 'suecia', 'russia', 'noruega', 'dinamarca'}
escandinavia = {'noruega', 'dinamarca', 'suecia'}

norte_europa_nao_escandinavo = norte_europa - escandinavia
print(norte_europa_nao_escandinavo)
```

```python
# A ordem dos operandos importa: aqui buscamos o oposto
escandinavo_nao_norte_europa = escandinavia - norte_europa
print(escandinavo_nao_norte_europa)  # set() vazio: todo país escandinavo está no norte da Europa
```

> 💡 O Python também oferece `|` (união, todos os itens dos dois conjuntos) e `&` (interseção, apenas os itens presentes em ambos). Aqui, diferente dos booleanos, `|` e `&` **são** a forma correta de usar com conjuntos.

### 2.4 Métodos

| Método | O que faz |
| --- | --- |
| `.add(valor)` | Adiciona um item ao conjunto |
| `.remove(valor)` | Remove um item do conjunto |

```python
cursos_disponiveis = {'Exatas', 'Humanas', 'Biológicas'}
print(cursos_disponiveis)
```

```python
# .add(): adiciona um item (se já existir, o conjunto não muda)
cursos_disponiveis.add('Saúde')
print(cursos_disponiveis)
```

```python
# .remove(): remove um item existente
cursos_disponiveis.remove('Saúde')
print(cursos_disponiveis)
```

### 2.5 Conversão

É muito comum converter uma **lista em conjunto** (para eliminar repetições) e, em seguida, o **conjunto de volta em lista** (para poder usar índices e métodos de lista novamente).

```python
times_paulistas = ['São Paulo', 'Palmeiras', 'Corinthians', 'Santos']

# list -> set: elimina eventuais repetições
times_sem_repeticao = set(times_paulistas)
print(times_sem_repeticao)
print(type(times_sem_repeticao))
```

```python
# set -> list: para voltar a usar índices e métodos de lista
times_lista = list(times_sem_repeticao)
print(times_lista)
print(type(times_lista))
```

### 2.6 Revisitando a motivação

Combinando as duas conversões, resolvemos o problema das hashtags em uma única linha: juntamos as listas, convertemos para `set` (removendo repetições) e voltamos para `list`.

```python
hashtags_semana_unicas = list(set(hashtags_segunda + hashtags_terca + hashtags_quarta + hashtags_quinta + hashtags_sexta))

print(hashtags_semana_unicas)
print(f'Total de hashtags únicas: {len(hashtags_semana_unicas)}')
```

## 3. Dicionários

Listas e conjuntos guardam valores "soltos", identificados apenas pela posição ou pela própria existência. Mas muitos dados do mundo real vêm em **pares**: um nome de rede e sua senha, um CPF e o nome do cliente. Para isso usamos os **dicionários**.

### 3.1 Motivação

Para se conectar a uma rede wi-fi, você precisa de duas informações: o **nome da rede** e a **senha de acesso**. Ao procurar por redes disponíveis, seu dispositivo encontra uma lista de nomes:

```python
redes_wifi_disponiveis = ['rede1', 'cnx_cnx', 'uai-fi', 'r3d3']
print(redes_wifi_disponiveis)
```

O problema é que essa lista só guarda os **nomes**. Não há como associar cada rede à sua senha correspondente. **Como armazenamos pares de informações relacionadas, como nome e senha?**

### 3.2 Definição

**Dicionários** armazenam dados no formato **chave-valor**: cada item é composto por uma **chave** (usada para buscar o dado) e um **valor** (o dado em si). São do tipo `dict` e são criados entre chaves `{}`, no formato `{chave: valor}`.

```python
brasil = {'capital': 'Brasília', 'idioma': 'Português', 'populacao_milhoes': 210}

print(brasil)
print(type(brasil))
```

> ⚠️ **Atenção:** não é permitido ter **chaves duplicadas**. Se você repetir uma chave, o último valor atribuído a ela **sobrescreve** os anteriores, sem gerar erro:

```python
carro = {
    'marca': 'Volkswagen',
    'modelo': 'Polo',
    'ano': 2021,
    'ano': 2004,  # sobrescreve o valor anterior da chave 'ano'
}

print(carro)
```

Um valor de um dicionário também pode ser **outro dicionário** — isso é chamado de **dicionário composto**, muito útil para representar dados com hierarquia:

```python
cadastro = {
    'andre': {
        'nome': 'André Perez',
        'ano_nascimento': 1992,
        'pai': {'nome': 'Carlos Perez', 'ano_nascimento': 1965},
        'mae': {'nome': 'Marta Perez', 'ano_nascimento': 1968},
    }
}

print(cadastro)
```

Para acessar um valor dentro de vários níveis, encadeamos os colchetes com cada chave, do nível mais externo para o mais interno:

```python
ano_nascimento_mae = cadastro['andre']['mae']['ano_nascimento']
print(ano_nascimento_mae)
```

### 3.3 Operações

Diferente das listas, os itens de um dicionário são acessados e atualizados pela **chave**, e não pela posição.

```python
score_credito = {'123': 750, '789': 980}
```

```python
# Acessando um valor pela chave
score_cliente_123 = score_credito['123']
score_cliente_789 = score_credito['789']

print(score_cliente_123)
print(score_cliente_789)
```

```python
# Atualizando um valor: atribuímos um novo valor à chave já existente
score_credito['123'] = 435
print(score_credito)
```

```python
# Adicionando um novo par chave-valor: basta atribuir a uma chave nova
score_credito['456'] = 1000
print(score_credito)
```

> ⚠️ **Atenção:** acessar uma chave que **não existe** com `dicionario[chave]` gera um erro `KeyError`. Para buscar com segurança, use o método `.get(chave)`, que retorna `None` (em vez de erro) quando a chave não existe.

```python
print(score_credito.get('456'))  # existe: retorna o valor
print(score_credito.get('999'))  # não existe: retorna None, sem erro
```

### 3.4 Métodos

| Método | O que faz |
| --- | --- |
| `.update(dict)` | Adiciona ou atualiza vários pares chave-valor de uma vez |
| `.pop(chave)` | Remove um item pela chave e **retorna** o valor removido |
| `.get(chave)` | Retorna o valor da chave, ou `None` se ela não existir |

```python
artigo = dict(
    titulo='Módulo 02 - Estruturas de Dados',
    corpo='Tópicos, Aulas, Listas, Conjuntos, Dicionários...',
    total_caracteres=1530,
)
print(artigo)
```

```python
# .update(): atualiza um ou mais pares de uma vez
artigo.update({'total_caracteres': 7850})
print(artigo)
```

```python
# .pop(): remove pela CHAVE e devolve o valor removido
total_caracteres = artigo.pop('total_caracteres')

print(f'Valor removido: {total_caracteres}')
print(artigo)
```

### 3.5 Conversão

Podemos converter as **chaves** e os **valores** de um dicionário em listas, com os métodos `.keys()` e `.values()`:

```python
artigo = dict(
    titulo='Módulo 02 - Estruturas de Dados',
    corpo='Tópicos, Aulas, Listas, Conjuntos, Dicionários...',
    total_caracteres=1530,
)
```

```python
chaves = list(artigo.keys())

print(chaves)
print(type(chaves))
```

```python
valores = list(artigo.values())

print(valores)
print(type(valores))
```

### 3.6 Revisitando a motivação

Agora podemos representar cada rede wi-fi como um dicionário com as chaves `nome` e `senha`, e guardar todas elas em uma lista:

```python
redes_wifi_disponiveis = []

rede = {'nome': 'rede1', 'senha': 'cnx_cnx'}
redes_wifi_disponiveis.append(rede)

rede = {'nome': 'uai-fi', 'senha': 'r3d3'}
redes_wifi_disponiveis.append(rede)

print(redes_wifi_disponiveis)
```

Agora sim: para cada rede da lista, sabemos exatamente qual é o nome e qual é a senha correspondente — o problema do início foi resolvido combinando **lista** (a coleção de redes) e **dicionário** (os dados de cada rede).

## Resumo do Módulo

| Estrutura | Tipo | Ordenada? | Repete valores? | Acesso | Principais métodos |
| --- | --- | --- | --- | --- | --- |
| Lista | `list` | Sim | Sim | Por índice `lista[i]` | `.append()`, `.insert()`, `.remove()`, `.pop()` |
| Conjunto | `set` | Não | Não | Não indexado | `.add()`, `.remove()`, `-`, `\|`, `&` |
| Dicionário | `dict` | Sim (desde o Python 3.7) | Chaves não, valores sim | Por chave `dic[chave]` | `.update()`, `.pop()`, `.get()`, `.keys()`, `.values()` |

Funções nativas vistas neste módulo: `list()`, `set()`, `dict()`, `sum()`, `len()`.
