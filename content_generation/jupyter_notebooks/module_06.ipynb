## 1. Um pouco de teoria

Antes de escrever código, vale entender a ideia por trás da orientação a objetos. Nesta seção vamos ver onde a POO se encaixa entre os estilos de programação e conhecer os três conceitos que sustentam o restante do módulo: **classe**, **objeto** e **herança**.

### 1.1 Paradigmas de programação

Um **paradigma de programação** é um **estilo** de organizar e escrever código. Os três mais conhecidos são:

| Paradigma | Ideia central | Pergunta que o código responde |
| --- | --- | --- |
| **Imperativo** | Uma sequência de instruções que alteram variáveis passo a passo | "Como fazer?" |
| **Funcional** | Composição de funções que recebem valores e devolvem resultados, sem alterar dados externos | "O que calcular?" |
| **Orientado a objetos** | Dados e comportamentos agrupados em **objetos**, que representam elementos do problema | "Quem faz o quê?" |

O Python é uma linguagem **multiparadigma**: ele permite usar os três estilos, inclusive misturados no mesmo programa. Para comparar, veja o mesmo problema resolvido nos três estilos: calcular o total de um carrinho de compras.

**Estilo imperativo:** percorremos os preços e acumulamos o total em uma variável.

````python
precos_carrinho = [49.90, 15.00, 120.50]

# Passo a passo: a variável total_imperativo é alterada a cada volta do laço
total_imperativo = 0
for preco in precos_carrinho:
    total_imperativo += preco

print(f'Total (imperativo): {total_imperativo:.2f}')
````

**Estilo funcional:** descrevemos o resultado combinando funções, sem alterar nenhuma variável no caminho.

````python
# sum() recebe a lista e devolve o total, sem modificar nada
total_funcional = sum(precos_carrinho)

print(f'Total (funcional): {total_funcional:.2f}')
````

**Estilo orientado a objetos:** criamos um "carrinho" que guarda os próprios itens e sabe calcular o próprio total. Não se preocupe com a sintaxe agora: é exatamente isso que vamos aprender neste módulo.

````python
class Carrinho:

    def __init__(self):
        self.precos = []

    def adicionar(self, preco: float) -> None:
        self.precos.append(preco)

    def total(self) -> float:
        return sum(self.precos)


# O carrinho guarda os dados E as ações que fazem sentido para ele
meu_carrinho = Carrinho()
meu_carrinho.adicionar(49.90)
meu_carrinho.adicionar(15.00)
meu_carrinho.adicionar(120.50)

print(f'Total (orientado a objetos): {meu_carrinho.total():.2f}')
````

> 💡 **Dica:** nenhum paradigma é "melhor" em todos os casos. Para um cálculo simples, uma linha como `sum(precos_carrinho)` resolve. A orientação a objetos brilha quando o sistema cresce e passa a ter muitos elementos (clientes, produtos, pedidos), cada um com seus dados e suas regras.

### 1.2 Classe

Uma **classe** é a representação de um elemento, real ou abstrato, e funciona como uma **receita** (ou um molde) para criar objetos. Ela define:

- quais **características** todo elemento daquele tipo tem (os **atributos**);
- quais **ações** ele pode realizar (os **métodos**).

**Exemplo:** a classe `Pessoa` define que toda pessoa tem nome, idade e documento, e que toda pessoa pode falar e dormir.

### 1.3 Objeto

Um **objeto** é uma **instância** de uma classe, ou seja, algo concreto criado a partir da receita. A classe descreve; o objeto existe e tem valores próprios.

**Exemplo:** André, de 30 anos, é uma instância da classe `Pessoa`. Maria, de 56 anos, é outra instância da mesma classe. Os dois seguem a mesma receita, mas cada um tem os seus dados.

Pense na planta de uma casa: a planta é a **classe**, e cada casa construída a partir dela é um **objeto**. Com uma única planta, construímos quantas casas quisermos, e cada uma pode ter uma cor diferente.

### 1.4 Herança

A **herança** é uma forma de **especializar** uma classe: criamos uma nova classe que aproveita tudo o que outra já tem e acrescenta o que é específico dela.

**Exemplo:** um estudante é um tipo especial de pessoa. Ele tem nome, idade e documento (como qualquer pessoa), mas também estuda em uma universidade.

| Conceito | Analogia | Exemplo |
| --- | --- | --- |
| Classe | A receita ou a planta | `Pessoa` |
| Objeto | O prato pronto ou a casa construída | André, Maria, Pedro |
| Herança | Uma variação da receita | `Estudante` é uma `Pessoa` |

## 2. Classes

As classes são o ponto de partida da orientação a objetos: é nelas que definimos quais dados e quais comportamentos um tipo de elemento terá. Nesta seção vamos construir, passo a passo, a classe `Pessoa`.

### 2.1 Motivação

Você trabalha no sistema de cadastro de um banco e precisa guardar os dados de vários clientes. Com o que vimos até agora, uma opção seria usar variáveis soltas:

````python
# Cada informação de cada cliente em uma variável diferente
nome_cliente_1 = 'Andre Perez'
idade_cliente_1 = 30

nome_cliente_2 = 'Maria Perez'
idade_cliente_2 = 56

print(f'{nome_cliente_1}, {idade_cliente_1} anos')
print(f'{nome_cliente_2}, {idade_cliente_2} anos')
````

Com 2 clientes funciona. Mas e com 2 mil? E quando cada cliente tiver 15 informações? E se toda regra do sistema (fazer aniversário, validar documento, exibir os dados) precisar receber todas essas variáveis como parâmetro? O código fica repetitivo e fácil de quebrar.

**Como podemos agrupar os dados de um cliente e as ações que fazem sentido para ele em uma única estrutura?**

### 2.2 Definição

Uma **classe** é uma receita para a criação de objetos. A estrutura geral é:

```python
class NomeDaClasse:

    def __init__(self, parametros):
        self.atributo_1 = ...
        self.atributo_2 = ...

    def metodo_1(self, parametros):
        ...

    def metodo_2(self, parametros):
        ...
```

Alguns pontos importantes dessa estrutura:

- A palavra reservada `class` inicia a definição da classe;
- Por convenção ([PEP 8](https://peps.python.org/pep-0008/#class-names)), nomes de classe usam **`PascalCase`**: cada palavra começa com letra maiúscula e não há `_` (ex: `Pessoa`, `ContaBancaria`, `ArquivoCSV`). Variáveis e funções continuam em `snake_case`;
- Tudo o que pertence à classe fica **indentado** dentro dela (4 espaços).

Vamos criar a versão mais simples possível da classe `Pessoa`. A palavra `pass` indica um bloco vazio: a classe existe, mas ainda não faz nada.

````python
class Pessoa:
    pass


print(Pessoa)
````

> 💡 **Dica:** em códigos mais antigos você vai encontrar `class Pessoa(object):`. No Python 3, toda classe já herda de `object` automaticamente (veremos herança na seção 4), então `class Pessoa:` é equivalente e é a forma recomendada hoje.

### 2.3 Atributos

Os **atributos** representam as **características** da classe. Eles são criados dentro de um método especial chamado **`__init__`** (de *initialize*, "inicializar"), que o Python executa automaticamente sempre que um novo objeto é criado. Ele é conhecido como o **construtor** da classe.

Dois detalhes que costumam confundir no começo:

| Elemento | O que é |
| --- | --- |
| `self` | Uma referência ao **próprio objeto** que está sendo criado. Todo método da classe recebe `self` como primeiro parâmetro |
| `self.nome = nome` | Cria o atributo `nome` **no objeto** e guarda nele o valor recebido no parâmetro `nome` |

````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str):
        # Cada self.atributo fica guardado dentro do objeto
        self.nome = nome
        self.idade = idade
        self.documento = documento
````

> 💡 **Dica:** as anotações `: str` e `: int` são **dicas de tipo** (*type hints*). Elas documentam que tipo de valor cada parâmetro espera e ajudam o editor a apontar erros, mas o Python **não as obriga**: são apenas documentação.

Assim como em funções, os parâmetros do `__init__` podem ter **valores padrão**. Nem toda pessoa informa o documento no cadastro, então vamos deixá-lo opcional, com o valor `None`:

````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento  # None quando não for informado
````

### 2.4 Métodos

Os **métodos** representam as **ações** da classe. Eles são funções definidas **dentro** da classe e, por isso, sempre recebem `self` como primeiro parâmetro. É pelo `self` que um método acessa (e altera) os atributos do próprio objeto.

Nesta versão, a `Pessoa` ganha três ações:

| Método | O que faz |
| --- | --- |
| `dormir(horas)` | Exibe o progresso do sono, hora a hora |
| `falar(texto)` | Exibe uma fala da pessoa |
| `fazer_aniversario()` | **Altera** o atributo `idade`, somando 1 |

````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento

    def dormir(self, horas: int) -> None:
        # self.nome: o método acessa um atributo do próprio objeto
        for hora in range(1, horas + 1):
            print(f'{self.nome} está dormindo... {hora}h de {horas}h')

    def falar(self, texto: str) -> None:
        print(f'{self.nome} diz: {texto}')

    def fazer_aniversario(self) -> None:
        # Um método também pode ALTERAR o estado do objeto
        self.idade += 1
        print(f'Parabéns, {self.nome}! Agora você tem {self.idade} anos.')
````

> 💡 **Dica:** a anotação `-> None` indica que o método **não retorna** nenhum valor (ele só exibe algo ou altera o objeto). Um método que devolve um texto seria anotado com `-> str`.

### 2.5 O método especial `__str__`

Métodos cujo nome começa e termina com dois *underlines* (como `__init__`) são **métodos especiais** (também chamados de *dunder methods*, de *double underscore*). O Python os chama automaticamente em situações específicas.

Um dos mais úteis é o **`__str__`**, chamado quando o objeto é convertido em texto, por exemplo, dentro de um `print()`. Veja o que acontece quando imprimimos um objeto da classe atual, que **ainda não** tem `__str__` (a criação de objetos será detalhada na seção 3):

````python
# Sem __str__, o Python mostra apenas a classe e o endereço de memória do objeto
pessoa_teste = Pessoa(nome='Andre Perez', idade=30)
print(pessoa_teste)
````

Pouco útil, não é? Vamos acrescentar o `__str__` à classe. Ele deve **retornar** (e não imprimir) uma *string* que descreva o objeto. Esta é a versão final da classe `Pessoa`, que vamos usar no restante do módulo:

````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento

    def dormir(self, horas: int) -> None:
        for hora in range(1, horas + 1):
            print(f'{self.nome} está dormindo... {hora}h de {horas}h')

    def falar(self, texto: str) -> None:
        print(f'{self.nome} diz: {texto}')

    def fazer_aniversario(self) -> None:
        self.idade += 1
        print(f'Parabéns, {self.nome}! Agora você tem {self.idade} anos.')

    def __str__(self) -> str:
        # Evita exibir "None" quando o documento não foi informado
        if self.documento is None:
            documento_exibido = 'não informado'
        else:
            documento_exibido = self.documento
        return f'{self.nome}, {self.idade} anos, documento: {documento_exibido}'
````

````python
# Agora o print() usa o __str__ e exibe uma descrição legível
pessoa_teste = Pessoa(nome='Andre Perez', idade=30)
print(pessoa_teste)
````

> ⚠️ **Atenção:** sempre que você altera uma classe, é preciso **executar de novo** a célula que a define **e** recriar os objetos. Objetos criados antes continuam seguindo a versão antiga da classe.

## 3. Objetos

Uma classe sozinha é só uma receita: quem de fato guarda dados e executa ações são os **objetos**. Nesta seção vamos criar objetos, acessar e alterar seus atributos, chamar seus métodos e aplicar tudo isso em um exemplo real de leitura de arquivos.

### 3.1 Definição

Um **objeto** é uma **instância** de uma classe. Para criá-lo, "chamamos" a classe como se fosse uma função, passando os parâmetros do `__init__` (sem o `self`, que o Python preenche sozinho). Em seguida, usamos o **ponto** (`.`) para acessar atributos e métodos:

```python
objeto = NomeDaClasse(parametros)

objeto.atributo    # acessa uma característica
objeto.metodo()    # executa uma ação
```

````python
# Três objetos criados a partir da mesma classe, cada um com seus próprios dados
andre = Pessoa(nome='Andre Perez', idade=30, documento='123')
maria = Pessoa(nome='Maria Perez', idade=56, documento='456')
pedro = Pessoa(nome='Pedro Perez', idade=8)  # sem documento: usa o valor padrão None

print(andre)
print(maria)
print(pedro)
````

Podemos confirmar a qual classe um objeto pertence com `type()` ou, de forma mais comum no dia a dia, com `isinstance()`, que responde com um booleano:

````python
print(type(andre))
print(isinstance(andre, Pessoa))
print(isinstance('Andre Perez', Pessoa))  # uma string não é uma Pessoa
````

### 3.2 Manipulando atributos

Um atributo funciona como uma variável que "mora" dentro do objeto: podemos **ler** o valor, usá-lo em cálculos e funções e também **alterá-lo**.

````python
# Lendo atributos
print(andre.nome)
print(andre.idade)
````

**Exemplo:** usando um atributo como argumento de uma função comum.

````python
def maior_de_idade(idade: int) -> bool:
    return idade >= 18


# Passamos apenas o atributo idade de cada objeto para a função
for pessoa in [andre, maria, pedro]:
    if maior_de_idade(idade=pessoa.idade):
        print(f'{pessoa.nome} é maior de idade')
    else:
        print(f'{pessoa.nome} é menor de idade')
````

**Exemplo:** consultando o *score* de crédito de um cliente a partir do documento. O documento do objeto serve de chave para buscar o valor em um dicionário:

````python
# Score de crédito de cada cliente, indexado pelo número do documento
score_credito = {'123': 750, '456': 812, '789': 327}

score_andre = score_credito[andre.documento]
print(f'Score de {andre.nome}: {score_andre}')
````

> ⚠️ **Atenção:** o `pedro` foi criado sem documento (`None`), então `score_credito[pedro.documento]` geraria um `KeyError`. Com o método `.get()` do dicionário, podemos definir um valor padrão para quando a chave não existir.

````python
# score_pedro = score_credito[pedro.documento]           # ❌ KeyError: None
score_pedro = score_credito.get(pedro.documento, 'sem score')  # ✅ valor padrão

print(f'Score de {pedro.nome}: {score_pedro}')
````

Também podemos **alterar** um atributo diretamente, com uma atribuição comum. A mudança vale apenas para aquele objeto:

````python
# Pedro trouxe o documento: atualizamos só o objeto pedro
pedro.documento = '789'

print(pedro)
print(andre)  # andre não foi afetado
````

### 3.3 Chamando métodos

Os métodos são chamados com a sintaxe `objeto.metodo(argumentos)`. Repare que **não** passamos o `self`: o Python entende que o `self` é o objeto que vem antes do ponto.

````python
# O self dentro de dormir() é o próprio andre
andre.dormir(horas=3)
````

````python
andre.falar(texto='Olá, pessoal!')
maria.falar(texto='Bom dia!')
````

O método `fazer_aniversario()` altera o estado do objeto. Observe a idade antes e depois:

````python
print(maria)

maria.fazer_aniversario()

print(maria)
````

> ⚠️ **Atenção:** para executar um método é obrigatório usar os parênteses. Escrever `maria.fazer_aniversario` (sem `()`) não gera erro, mas também **não executa** nada: apenas faz referência ao método.

### 3.4 Revisitando a motivação

Lembra do cadastro de clientes da seção 2.1? Com a classe `Pessoa`, cada cliente vira **um único objeto** que carrega seus dados e suas ações. Guardando os objetos em uma lista, cadastrar 3 ou 3 mil clientes exige o mesmo código:

````python
clientes = [
    Pessoa(nome='Andre Perez', idade=30, documento='123'),
    Pessoa(nome='Maria Perez', idade=56, documento='456'),
    Pessoa(nome='Pedro Perez', idade=8),
]

# Uma única regra aplicada a todos os clientes
for cliente in clientes:
    print(cliente)
````

````python
# Virada de ano: todos os clientes fazem aniversário
for cliente in clientes:
    cliente.fazer_aniversario()
````

Nenhuma variável solta, nenhuma função recebendo dezenas de parâmetros: cada objeto sabe cuidar dos próprios dados.

### 3.5 Tudo no Python é um objeto

Você usa objetos desde o primeiro módulo, talvez sem perceber. Números, textos, listas, dicionários e até funções são objetos de alguma classe nativa do Python. A função `type()` mostra a classe de cada um:

````python
tipos = [
    type(10),
    type(1.1),
    type('Ready To Deploy'),
    type(True),
    type(None),
    type([1, 2, 3]),
    type({1, 2, 3}),
    type({'janeiro': 1}),
    type(lambda x: x),
]

for tipo in tipos:
    print(tipo)
````

Repare no `<class '...'>` em cada linha: `int`, `str`, `list` e os demais são **classes**. Por isso eles têm métodos, que chamamos exatamente como fizemos com a `Pessoa`:

````python
# nome é um objeto da classe str, e upper() é um método dessa classe
nome = 'Andre Perez'
print(nome.upper())
````

````python
# juros é um objeto da classe list, e pop() é um método dessa classe
juros = [0.02, 0.07, 0.15]

juro_removido = juros.pop(1)  # remove e retorna o item da posição 1
print(juro_removido)
print(juros)
````

### 3.6 Revisão: `strip()` e `split()`

No próximo exemplo vamos ler um arquivo CSV linha a linha. Para isso, dois métodos de *string* serão fundamentais:

| Método | O que faz | Exemplo | Resultado |
| --- | --- | --- | --- |
| `.strip()` | Remove espaços e quebras de linha (`\n`) do início e do fim | `' banana \n'.strip()` | `'banana'` |
| `.strip(caracteres)` | Remove os caracteres informados do início e do fim | `'..banana..'.strip('.')` | `'banana'` |
| `.split(separador)` | Divide o texto em uma **lista** a cada separador | `'a,b,c'.split(',')` | `['a', 'b', 'c']` |

````python
# strip() sem argumentos: remove espaços e quebras de linha das pontas
fruta = ' banana \n'
print(fruta.strip())
````

````python
# strip(caracteres): remove QUALQUER um dos caracteres informados, só nas pontas
fruta = ',,,,,rrttgg.....banana....rrr'
print(fruta.strip(',.grt'))
````

````python
# Combinando os dois: primeiro limpamos as pontas, depois dividimos
frase = ' O gato pulou sobre o muro '
palavras = frase.strip().split()  # sem argumento, split() divide nos espaços
print(palavras)
````

````python
# Uma linha típica de arquivo CSV: valores separados por vírgula e um \n no final
linha_csv = '30,unemployed,married,primary\n'
valores = linha_csv.strip().split(',')
print(valores)
````

### 3.7 Exemplo prático: classe `ArquivoCSV`

Arquivos **CSV** (*Comma-Separated Values*) são tabelas em formato de texto: cada linha é um registro e os valores são separados por vírgula. A primeira linha costuma trazer o nome das colunas.

Vamos criar uma classe que lê um CSV com dados de clientes de um banco e permite extrair qualquer coluna. Primeiro, criamos o arquivo `banco.csv`:

````python
conteudo_csv = '''age,job,marital,education,default,balance,housing,loan
30,unemployed,married,primary,no,1787,no,no
33,services,married,secondary,no,4789,yes,yes
35,management,single,tertiary,no,1350,yes,no
30,management,married,tertiary,no,1476,yes,yes
59,blue-collar,married,secondary,no,0,yes,no
35,management,single,tertiary,no,747,no,no
36,self-employed,married,tertiary,no,307,yes,no
39,technician,married,secondary,no,147,yes,no
41,entrepreneur,married,tertiary,no,221,yes,no
43,services,married,primary,no,-88,yes,yes'''

# Gravando o texto em um arquivo no ambiente atual
with open('banco.csv', mode='w', encoding='utf8') as arquivo:
    arquivo.write(conteudo_csv)
````

> 💡 **Dica:** no Colab você também encontrará o comando mágico `%%writefile banco.csv` na primeira linha de uma célula, que grava o conteúdo da célula em um arquivo. Aqui usamos `open()` porque funciona em qualquer ambiente Python.

Agora, a classe. Ela terá dois tipos de método:

| Método | Tipo | O que faz |
| --- | --- | --- |
| `_extrair_conteudo()` | Interno | Lê o arquivo e retorna a lista de linhas |
| `_extrair_nome_colunas()` | Interno | Retorna os nomes das colunas (a primeira linha) |
| `extrair_coluna(indice_coluna)` | Público | Retorna os valores de uma coluna a partir da sua posição |
| `extrair_coluna_por_nome(nome_coluna)` | Público | Retorna os valores de uma coluna a partir do seu nome |

O `_` no início do nome é uma **convenção** do Python para indicar que um método é **de uso interno** da classe: ele existe para ajudar os outros métodos, e não para ser chamado de fora.

````python
class ArquivoCSV:

    def __init__(self, caminho_arquivo: str):
        self.caminho_arquivo = caminho_arquivo
        # O próprio construtor já lê o arquivo e descobre as colunas
        self.conteudo = self._extrair_conteudo()
        self.colunas = self._extrair_nome_colunas()

    def _extrair_conteudo(self) -> list:
        with open(self.caminho_arquivo, mode='r', encoding='utf8') as arquivo:
            return arquivo.readlines()

    def _extrair_nome_colunas(self) -> list:
        # A primeira linha do arquivo é o cabeçalho
        return self.conteudo[0].strip().split(',')

    def extrair_coluna(self, indice_coluna: int) -> list:
        coluna = []
        # [1:] pula a primeira linha, pois o cabeçalho não é um dado
        for linha in self.conteudo[1:]:
            valores_linha = linha.strip().split(',')
            coluna.append(valores_linha[indice_coluna])
        return coluna

    def extrair_coluna_por_nome(self, nome_coluna: str) -> list:
        # Um método pode chamar outro método do mesmo objeto via self
        indice_coluna = self.colunas.index(nome_coluna)
        return self.extrair_coluna(indice_coluna=indice_coluna)
````

Ao criar o objeto, o `__init__` já lê o arquivo e preenche os atributos `conteudo` e `colunas`:

````python
arquivo_banco = ArquivoCSV(caminho_arquivo='banco.csv')

print(arquivo_banco.colunas)
````

````python
# Cada item do atributo conteudo é uma linha do arquivo (repare no \n no final)
print(arquivo_banco.conteudo[:3])
````

**Exemplo:** extraindo a coluna `job` (profissão), que está na posição `1`.

````python
profissoes = arquivo_banco.extrair_coluna(indice_coluna=1)
print(profissoes)
````

**Exemplo:** extraindo a coluna `education` (escolaridade), que está na posição `3`.

````python
escolaridades = arquivo_banco.extrair_coluna(indice_coluna=3)
print(escolaridades)
````

**Exemplo:** decorar a posição de cada coluna é trabalhoso e fácil de errar. Com `extrair_coluna_por_nome()`, basta informar o nome:

````python
estados_civis = arquivo_banco.extrair_coluna_por_nome(nome_coluna='marital')
print(estados_civis)

# O resultado é uma lista comum, então podemos usar os métodos de lista
print(f"Clientes casados: {estados_civis.count('married')}")
````

**Exemplo:** os valores lidos de um arquivo de texto são sempre *strings*. Para fazer contas, convertemos cada valor. Aqui calculamos o saldo médio (`balance`) dos clientes:

````python
saldos_texto = arquivo_banco.extrair_coluna_por_nome(nome_coluna='balance')

# Convertendo cada saldo de str para int
saldos = []
for saldo in saldos_texto:
    saldos.append(int(saldo))

saldo_medio = sum(saldos) / len(saldos)
print(f'Saldo médio: {saldo_medio:.2f}')
````

> ⚠️ **Atenção:** o Python **não impede** que você chame `arquivo_banco._extrair_conteudo()` de fora da classe, mas isso vai contra a convenção do `_`. Métodos internos podem mudar ou deixar de existir em versões futuras da classe; use sempre os atributos e métodos públicos.

> 💡 **Dica:** esta classe é ótima para entender orientação a objetos, mas no dia a dia existem ferramentas prontas para ler CSVs, como o módulo nativo [`csv`](https://docs.python.org/pt-br/3/library/csv.html) e a biblioteca **Pandas**, que veremos em módulos futuros. Ambas são construídas justamente com classes e objetos.

## 4. Herança

Todo estudante é uma pessoa, mas nem toda pessoa é um estudante. A **herança** permite que a classe `Estudante` reaproveite tudo o que a classe `Pessoa` já tem, sem copiar código, e acrescente apenas o que é específico de um estudante.

### 4.1 Definição

A herança cria uma **especialização** de uma classe. A classe original é chamada de **classe mãe** (ou superclasse), e a nova, de **classe filha** (ou subclasse). A classe filha herda todos os atributos e métodos da mãe. A estrutura geral é:

```python
class ClasseMae:

    def __init__(self, parametros):
        ...

class ClasseFilha(ClasseMae):

    def __init__(self, parametros):
        super().__init__(parametros_da_mae)
        self.atributo_3 = ...
        self.atributo_4 = ...

    def metodo_3(self, parametros):
        ...
```

| Elemento | O que faz |
| --- | --- |
| `class ClasseFilha(ClasseMae):` | O nome da classe mãe entre parênteses indica a herança |
| `super()` | Dá acesso aos métodos da classe mãe |
| `super().__init__(...)` | Executa o construtor da mãe, que cria os atributos herdados |

> 💡 **Dica:** uma boa forma de saber se a herança faz sentido é a pergunta **"é um?"**: um estudante **é uma** pessoa? Sim, então `Estudante` pode herdar de `Pessoa`.

### 4.2 Criando a classe `Estudante`

Vamos reaproveitar a classe `Pessoa` da seção 2.5. Todo estudante estuda em uma universidade, então primeiro criamos a classe `Universidade`:

````python
class Universidade:

    def __init__(self, nome: str):
        self.nome = nome
````

Agora especializamos a classe `Pessoa` na classe `Estudante`. O construtor recebe os dados de uma pessoa **mais** a universidade, repassa os dados de pessoa para o construtor da mãe com `super()` e cria o único atributo novo:

````python
class Estudante(Pessoa):

    def __init__(self, nome: str, idade: int, documento: str, universidade: Universidade):
        # A classe mãe (Pessoa) cria os atributos nome, idade e documento
        super().__init__(nome=nome, idade=idade, documento=documento)
        # Atributo exclusivo de Estudante
        self.universidade = universidade
````

> ⚠️ **Atenção:** o `super().__init__()` é obrigatório quando a classe filha tem o próprio `__init__`. Sem ele, o construtor da mãe **não é executado**, e os atributos `nome`, `idade` e `documento` simplesmente não existem no objeto, gerando um `AttributeError` na primeira vez em que forem usados.

Repare em outro detalhe: o atributo `universidade` guarda um **objeto** da classe `Universidade`. Isso é diferente de herança:

| Relação | Pergunta | Exemplo | Como aparece no código |
| --- | --- | --- | --- |
| Herança | "é um?" | Um estudante **é uma** pessoa | `class Estudante(Pessoa):` |
| Composição | "tem um?" | Um estudante **tem uma** universidade | `self.universidade = universidade` |

### 4.3 Manipulação

Criamos os objetos da mesma forma que antes. Primeiro a universidade, depois o estudante, que recebe o objeto da universidade como parâmetro:

````python
usp = Universidade(nome='Universidade de São Paulo')
andre_estudante = Estudante(nome='Andre Perez', idade=30, documento='123', universidade=usp)
````

A classe `Estudante` não define `__str__`, `falar()` nem `dormir()`, mas o objeto consegue usar todos eles, porque foram **herdados** de `Pessoa`:

````python
# __str__ herdado de Pessoa
print(andre_estudante)

# Métodos herdados de Pessoa
andre_estudante.falar(texto='Tenho prova amanhã!')
andre_estudante.dormir(horas=2)
````

Para acessar um atributo de um objeto que está **dentro** de outro, encadeamos os pontos:

````python
# andre_estudante.universidade é um objeto Universidade; .nome é o atributo dele
print(andre_estudante.universidade.nome)
````

Um estudante é, ao mesmo tempo, um `Estudante` **e** uma `Pessoa`. O `isinstance()` confirma isso, e o `issubclass()` verifica a relação entre as classes:

````python
print(isinstance(andre_estudante, Estudante))
print(isinstance(andre_estudante, Pessoa))  # True: todo Estudante é uma Pessoa
print(isinstance(andre, Estudante))         # False: nem toda Pessoa é um Estudante

print(issubclass(Estudante, Pessoa))
````

### 4.4 Novos métodos e sobrescrita

Além de herdar, a classe filha pode:

- **Acrescentar** métodos que só fazem sentido para ela (ex: `estudar()`);
- **Sobrescrever** (*override*) um método herdado, definindo outra versão com o mesmo nome. Quando o método é chamado, o Python usa a versão da classe filha.

Vamos melhorar o `Estudante`: o `__str__` herdado não menciona a universidade. Na nova versão, aproveitamos o texto da mãe com `super().__str__()` e acrescentamos o que falta:

````python
class Estudante(Pessoa):

    def __init__(self, nome: str, idade: int, documento: str, universidade: Universidade):
        super().__init__(nome=nome, idade=idade, documento=documento)
        self.universidade = universidade

    def estudar(self, materia: str) -> None:
        # Método novo, exclusivo de Estudante
        print(f'{self.nome} está estudando {materia} na {self.universidade.nome}')

    def __str__(self) -> str:
        # Sobrescrita: reaproveita o texto da Pessoa e acrescenta a universidade
        return f'{super().__str__()}, estudante da {self.universidade.nome}'
````

Como a classe foi redefinida, recriamos o objeto (lembre-se do aviso da seção 2.5):

````python
andre_estudante = Estudante(nome='Andre Perez', idade=30, documento='123', universidade=usp)

print(andre_estudante)  # usa o __str__ de Estudante
print(andre)            # continua usando o __str__ de Pessoa

andre_estudante.estudar(materia='Programação Orientada a Objetos')
````

> 💡 **Dica:** com herança, uma melhoria feita na classe mãe chega automaticamente a todas as classes filhas. Se amanhã a `Pessoa` ganhar um método `apresentar()`, todo `Estudante` passa a tê-lo também, sem nenhuma linha a mais.

## Resumo do Módulo

| Conceito | O que é | Sintaxe |
| --- | --- | --- |
| Classe | Receita para criar objetos | `class Pessoa:` |
| Construtor | Método executado ao criar o objeto; define os atributos | `def __init__(self, nome):` |
| `self` | Referência ao próprio objeto dentro dos métodos | `self.nome = nome` |
| Atributo | Característica do objeto | `andre.nome` |
| Método | Ação do objeto | `andre.falar(texto='Olá')` |
| `__str__` | Método especial que define o texto exibido pelo `print()` | `def __str__(self) -> str:` |
| Objeto | Instância concreta de uma classe | `andre = Pessoa(nome='Andre', idade=30)` |
| Método interno | Convenção para métodos de uso interno da classe | `def _extrair_conteudo(self):` |
| Herança | Classe filha que especializa uma classe mãe ("é um") | `class Estudante(Pessoa):` |
| `super()` | Acesso aos métodos da classe mãe | `super().__init__(nome=nome)` |
| Sobrescrita | Nova versão de um método herdado | redefinir `__str__` na classe filha |

Funções nativas vistas neste módulo: `type()`, `isinstance()`, `issubclass()`, `open()` e `sum()`.
