## 1. Installare Python

Per eseguire uno *script* fuori da Colab — sul proprio computer, oppure su un server — è necessario avere Python (e **PIP**, il suo gestore di pacchetti) installati sulla macchina che eseguirà il codice.


### 1.1 Su Google Colab

Python e PIP **sono già installati** sulla macchina virtuale che Colab mette a disposizione — non c'è bisogno di installare nulla.

**Esempio:** verifica della versione di Python dal Python stesso, con il modulo nativo `platform`.


````python
import platform

print(platform.python_version())
````

````python
print(platform.system())
````

> 💡 **Suggerimento:** la macchina virtuale di Colab esegue **Linux**, quindi il risultato del codice sopra sarà normalmente `Linux` — anche se il tuo computer è Windows o macOS. È un ambiente completamente separato dal tuo computer.


Oltre a eseguire codice Python in una cella, Colab permette di eseguire **comandi da terminale**: basta mettere un `!` prima del comando.

**Esempio:** verifica delle versioni di Python e PIP direttamente dal terminale di Colab.


````python
!python -V
````

````python
!pip -V
````

> 💡 Il `!` non è Python: dice a Colab "esegui questo come un comando da terminale (Bash)", e non come codice Python. Per questo motivo queste ultime due celle non vengono eseguite dal verificatore automatico di questo corso — funzionano solo all'interno di un notebook reale, come Colab.


### 1.2 Sul tuo computer locale

Prima di installare, verifica che Python non sia già installato. Apri il terminale (Windows: `cmd` o `PowerShell`; macOS/Linux: `terminale`) ed esegui:

```shell
python -V
pip -V
```

> ⚠️ **Attenzione:** su Windows, il comando è di solito `python`. Su macOS e Linux, è comune che sia `python3` (il nome `python` da solo potrebbe non esistere, oppure puntare a una vecchia installazione di Python 2). Se `python -V` non funziona, prova con `python3 -V`.

Se i comandi non vengono riconosciuti, Python non è installato. Vai sul [sito ufficiale di Python](https://www.python.org/), scarica l'installer per il tuo sistema operativo e segui le istruzioni — questo [tutorial di Real Python](https://realpython.com/installing-python/) è un'ottima guida passo passo.

> ⚠️ **Attenzione (Windows):** durante l'installazione, seleziona l'opzione **"Add Python to PATH"**. Senza di essa, il terminale non riconoscerà il comando `python`, anche con Python già installato — uno degli errori di configurazione più comuni tra i principianti.

Dopo l'installazione, chiudi e riapri il terminale, e lancia di nuovo `python -V` per confermare.

## 2. Eseguire codice nel terminale

Oltre a eseguire un intero file di script, Python ha una **modalità interattiva**, chiamata **REPL** (*Read-Eval-Print Loop*): scrivi una riga di codice, Python la esegue e mostra subito il risultato, una riga alla volta.


### 2.1 Su Google Colab

Ogni cella di codice di Colab funziona già come un REPL: scrivi un'istruzione, la esegui (`Shift + Invio`) e vedi il risultato immediatamente, senza dover eseguire un intero file `.py`. Per questo Colab è così adatto per imparare e testare codice rapidamente.


### 2.2 Sul tuo computer locale

Per aprire il REPL di Python sul tuo computer:

1. Crea una cartella per i file del corso (consigliato);
2. Apri il terminale **dentro questa cartella** (su Windows, `Shift + tasto destro` sulla cartella; su macOS/Linux, tasto destro e "Apri terminale qui");
3. Esegui il comando:

```shell
python
```

Il terminale cambia mostrando il *prompt* `>>>`, che indica che sei entrato in Python. Da qui in poi, ogni riga digitata viene eseguita immediatamente:

```python
>>> print("Ciao, mondo!")
Ciao, mondo!
>>> 2 + 2
4
```

Per uscire dal REPL, digita `quit()` (oppure `exit()`) e premi Invio, oppure chiudi il terminale.

## 3. Creare file di script

Un **file di script** è un file di testo con estensione `.py` contenente più righe di codice Python, salvate insieme per essere eseguite (e rieseguite) tutte insieme, sempre allo stesso modo. È il formato usato per le automazioni reali — qualcosa che deve girare da solo, senza un notebook aperto.


### 3.1 Su Google Colab

Colab non ha un editor di file tradizionale, ma puoi creare un file `.py` direttamente da una cella con il comando magico `%%writefile nome_del_file.py`: tutto ciò che viene **dopo** questa riga, nella stessa cella, viene salvato nel file.

**Esempio:** uno *script* che riceve un nome dalla riga di comando e stampa un saluto.


````python
%%writefile saudacao.py
import sys

nome = sys.argv[1]
print(f'Ciao, {nome}!')
````

Il modulo nativo `sys` memorizza, nella lista `sys.argv`, gli **argomenti da riga di comando**: `sys.argv[0]` è sempre il nome dello script stesso, e `sys.argv[1]`, `sys.argv[2]`... sono i valori aggiuntivi indicati da chi ha eseguito il comando.

Ora eseguiamo il file appena creato, passando `'Maria'` come argomento:


````python
!python saudacao.py Maria
````

> ⚠️ **Attenzione:** se lo script viene eseguito senza l'argomento atteso (`!python saudacao.py`, senza alcun nome), `sys.argv` avrà un solo elemento (`sys.argv[0]`), e provare ad accedere a `sys.argv[1]` genera un `IndexError`. Gli script per uso reale di solito verificano se gli argomenti necessari sono stati forniti prima di usarli.


**Esempio (mondo reale):** uno *script* di automazione che consulta un tasso d'interesse su un sito e registra il risultato, con data e ora, in un file `.csv` — il tipo di attività che di solito gira da sola, a un orario programmato (es. tutti i giorni alle 9:00).

> 💡 Questo è il **contenuto** di un file `extrair_taxa.py`, mostrato qui solo per lettura — non viene eseguito da questo notebook, perché dipende dall'accesso a internet e da un servizio esterno che può cambiare nel tempo. È esattamente questo tipo di attività, però, che ha senso far girare come *script*, e non dentro un notebook.

```python
import os
import json
from datetime import datetime

import requests

URL = 'https://exemplo-api-financeira.com/taxa-cdi'
ARQUIVO_SAIDA = 'taxa-cdi.csv'

data_e_hora = datetime.now()
data = data_e_hora.strftime('%Y-%m-%d')
hora = data_e_hora.strftime('%H:%M:%S')

# Consultazione del tasso sul sito
try:
    resposta = requests.get(URL)
    resposta.raise_for_status()
except requests.HTTPError:
    print('Tasso non trovato, ma lo script continuerà.')
    taxa_cdi = None
except Exception as erro:
    print('Errore di connessione, interruzione dello script.')
    raise erro
else:
    dado = json.loads(resposta.text)
    taxa_cdi = float(dado['taxa'].replace(',', '.'))

# Creazione del file con intestazione, se non esiste ancora
if not os.path.exists(ARQUIVO_SAIDA):
    with open(file=ARQUIVO_SAIDA, mode='w', encoding='utf8') as arquivo:
        arquivo.write('data,hora,taxa\n')

# Registra la riga solo se il tasso è stato ottenuto con successo
if taxa_cdi is not None:
    with open(file=ARQUIVO_SAIDA, mode='a', encoding='utf8') as arquivo:
        arquivo.write(f'{data},{hora},{taxa_cdi}\n')
    print('Successo!')
```

> 💡 **Suggerimento:** le API dei siti cambiano nel tempo (indirizzo, formato della risposta ecc.). Prima di usare un'API esterna in un progetto reale, controlla sempre la documentazione ufficiale e aggiornata del servizio.


### 3.2 Sul tuo computer locale

La dinamica è la stessa, cambia solo lo strumento usato per creare il file:

1. Crea il file `.py` con un editor di testo o di codice (nella prossima sezione vedremo le opzioni migliori);
2. Salva il file nella cartella in cui hai aperto il terminale;
3. Esegui con lo stesso comando usato su Colab, senza il `!` (che esiste solo dentro il notebook):

```shell
python saudacao.py Maria
```


### 3.3 Strumenti di sviluppo (IDE)

Un semplice editor di testo funziona per scrivere `.py`, ma una **IDE** (*Integrated Development Environment*, Ambiente di Sviluppo Integrato) rende il lavoro molto più produttivo: evidenzia gli errori di sintassi prima che tu esegua il codice, suggerisce l'autocompletamento, permette di eseguire il debug del codice passo passo e include già un terminale integrato.

Le due opzioni più usate nel settore sono:

- **Visual Studio Code**, di Microsoft, gratuito e leggero — la raccomandazione più comune per iniziare ([link](https://code.visualstudio.com/));
- **PyCharm**, di JetBrains, più robusto e focalizzato specificamente su Python, con una versione gratuita (*Community*) e una a pagamento (*Professional*) ([link](https://www.jetbrains.com/)).

> 💡 **Suggerimento:** lo stesso Visual Studio Code ha un'estensione ufficiale Jupyter/Colab, che permette di aprire ed eseguire notebook `.ipynb` (come quelli di questo corso) senza uscire dall'editor.
