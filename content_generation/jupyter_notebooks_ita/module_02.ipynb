## 1. Liste

Quando dobbiamo salvare **diverse informazioni dello stesso tipo di evento** — come le transazioni di un giorno — creare una variabile per ciascuna non è scalabile. Le **liste** risolvono esattamente questo problema.

### 1.1 Motivazione

L'app della tua banca registra tutti i tuoi movimenti finanziari. Alla fine della giornata, l'app somma tutte le transazioni al saldo iniziale per calcolare il **saldo finale**. Salvando ogni transazione in una variabile separata, il codice diventa così:

```python
saldo_iniziale_giorno_11 = 1000

transazione_1 = 243
transazione_2 = -798.58
transazione_3 = 427.12
transazione_4 = -10.91
```

```python
saldo_finale_giorno_11 = saldo_iniziale_giorno_11 + transazione_1 + transazione_2 + transazione_3 + transazione_4
print(round(saldo_finale_giorno_11, 2))
```

Funziona, ma nota il problema: se il giorno avesse 50 transazioni, servirebbero 50 variabili diverse, e il calcolo del saldo avrebbe 50 termini scritti a mano. **Esiste un modo migliore per memorizzare queste transazioni?**

### 1.2 Definizione

Le **liste** memorizzano sequenze **ordinate** e **mutabili** (possono essere modificate dopo la creazione) di valori, che possono ripetersi. Sono del tipo `list` e vengono create tra parentesi quadre `[]`, con gli elementi separati da virgola.

```python
dati_utente_web = ['André Perez', 'andre.perez', 'andre123', 'andre.perez@gmail.com']

print(dati_utente_web)
print(type(dati_utente_web))
```

Una lista può mescolare valori di **tipi diversi** — numeri, testi, booleani e persino altre liste:

```python
eta = 20
saldo_conto = 723.15
utente_connesso = True

dati_utente_web = ['André Perez', eta, 'andre.perez', 'andre.perez@gmail.com', saldo_conto, utente_connesso]

print(dati_utente_web)
print(type(dati_utente_web))
```

### 1.3 Operazioni

**Concatenazione (`+`)**: unisce due liste in una sola, mantenendo l'ordine.

**Esempio:** unire produttori di *hardware* mobile per regione.

```python
produttori_cina = ['xiaomi', 'huawei']
produttori_usa = ['apple', 'motorola']

produttori_mobile = produttori_cina + produttori_usa
print(produttori_mobile)
```

**Slicing**: come per le *stringhe*, accediamo agli elementi di una lista tramite la posizione (indice), che inizia da **0**. Gli indici negativi contano a partire dalla fine.

```python
print(f'Posizione 0:  {produttori_mobile[0]}')
print(f'Posizione -1: {produttori_mobile[-1]}')
```

Possiamo anche estrarre un **intervallo** di elementi con `lista[inizio:fine]` — l'elemento in posizione `fine` non viene incluso:

```python
produttori_cina = produttori_mobile[0:2]
produttori_usa = produttori_mobile[2:len(produttori_mobile)]

print(f'Cina: {produttori_cina}')
print(f'USA:  {produttori_usa}')
```

A differenza delle *stringhe*, le liste sono **mutabili**: possiamo modificare un elemento esistente assegnando un nuovo valore alla sua posizione.

```python
print(produttori_mobile)

produttori_mobile[2] = 'nokia'
print(produttori_mobile)
```

### 1.4 Metodi

I metodi sotto sono i più usati quotidianamente per aggiungere e rimuovere elementi da una lista:

| Metodo | Cosa fa |
| --- | --- |
| `.append(valore)` | Aggiunge un elemento alla **fine** della lista |
| `.insert(indice, valore)` | Inserisce un elemento in una **posizione specifica** |
| `.remove(valore)` | Rimuove la **prima occorrenza** di un valore |
| `.pop(indice)` | Rimuove l'elemento da una posizione e **restituisce** il valore rimosso |

```python
interessi_mensili = [0.05, 0.07, 0.02, 0.04, 0.08]
print(interessi_mensili)
```

```python
# .append(): aggiunge alla fine
interessi_mensili.append(0.09)
print(interessi_mensili)
```

```python
# .insert(): aggiunge in una posizione specifica, spostando gli altri elementi
interessi_mensili.insert(0, 0.10)
print(interessi_mensili)
```

```python
# .remove(): rimuove per VALORE (la prima occorrenza)
interessi_mensili.remove(0.10)
print(interessi_mensili)
```

```python
# .pop(): rimuove per INDICE e restituisce il valore rimosso
terzo_interesse = interessi_mensili.pop(2)
print(f'Valore rimosso: {terzo_interesse}')
print(interessi_mensili)
```

> 💡 **Suggerimento:** `.pop()` senza argomento rimuove e restituisce l'**ultimo** elemento della lista — utile per simulare una pila di elementi.

### 1.5 Conversione

Alcuni tipi possono essere convertiti in lista con la funzione `list()`. Una *stringa*, per esempio, diventa una lista con un elemento per ogni carattere:

```python
email = 'andre.perez@gmail.com'
caratteri_email = list(email)

print(email)
print(caratteri_email)
```

### 1.6 Rivisitando la motivazione

Ora possiamo salvare le transazioni del giorno in un'**unica lista**, e usare `.append()` per aggiungere ogni nuova transazione:

```python
transazioni_giorno_11 = []

transazioni_giorno_11.append(243)
transazioni_giorno_11.append(-798.58)
transazioni_giorno_11.append(427.12)
transazioni_giorno_11.append(-10.91)

print(transazioni_giorno_11)
```

E la funzione nativa `sum()` somma tutti gli elementi di una lista numerica in un colpo solo, senza dover sommare elemento per elemento:

```python
saldo_finale_giorno_11 = saldo_iniziale_giorno_11 + sum(transazioni_giorno_11)
print(round(saldo_finale_giorno_11, 2))
```

> 💡 Non importa se il giorno ha 4 o 400 transazioni: il codice resta esattamente lo stesso, perché `sum()` somma automaticamente tutti gli elementi della lista.

## 2. Insiemi

Le liste conservano elementi ripetuti e nell'ordine in cui sono stati inseriti. Ma a volte ciò che conta non è l'ordine né la ripetizione — è solo sapere **quali valori diversi** esistono. Per questo usiamo gli **insiemi**.

### 2.1 Motivazione

Lavori come analista di social media e devi scoprire **tutti gli hashtag diversi** entrati nei *trending topic* di un social network durante la settimana. Hai già gli hashtag di ogni giorno, in liste separate:

```python
hashtag_lunedi = ['#tiago', '#joao', '#bbb']
hashtag_martedi = ['#sarah', '#bbb', '#fiuk']
hashtag_mercoledi = ['#gil', '#thelma', '#lourdes']
hashtag_giovedi = ['#rafa', '#fora', '#danilo']
hashtag_venerdi = ['#juliete', '#arthur', '#bbb']
```

Se semplicemente concateniamo le liste della settimana, l'hashtag `#bbb` (che è apparso in tre giorni diversi) si ripeterà nel risultato:

```python
hashtag_settimana = hashtag_lunedi + hashtag_martedi + hashtag_mercoledi + hashtag_giovedi + hashtag_venerdi

print(hashtag_settimana)
print(f'Totale: {len(hashtag_settimana)}')
```

Nota che il totale (15) non rappresenta la quantità di hashtag **diversi** — `#bbb` è stato contato tre volte. **Come eliminiamo le ripetizioni?**

### 2.2 Definizione

Gli **insiemi** memorizzano sequenze **non ordinate** di valori, **senza ripetizioni**. Sono del tipo `set` e vengono creati tra parentesi graffe `{}`.

```python
frutti = {'banana', 'mela', 'uva', 'uva'}

print(frutti)
print(type(frutti))
```

Nota che `'uva'` è stata scritta due volte, ma compare **una sola volta** nell'insieme. Poiché gli insiemi non hanno ordine, la sequenza mostrata può variare a ogni esecuzione.

> ⚠️ **Attenzione:** non avendo un ordine, gli insiemi **non possono essere acceduti per indice** (`frutti[0]` genera un errore `TypeError`).

### 2.3 Operazioni

**Differenza (`-`)**: restituisce gli elementi che sono nel primo insieme, ma **non** sono nel secondo.

**Esempio:** paesi del nord Europa che non sono scandinavi.

```python
nord_europa = {'regno unito', 'svezia', 'russia', 'norvegia', 'danimarca'}
scandinavia = {'norvegia', 'danimarca', 'svezia'}

nord_europa_non_scandinavo = nord_europa - scandinavia
print(nord_europa_non_scandinavo)
```

```python
# L'ordine degli operandi conta: qui cerchiamo l'opposto
scandinavo_non_nord_europa = scandinavia - nord_europa
print(scandinavo_non_nord_europa)  # set() vuoto: ogni paese scandinavo è nel nord Europa
```

> 💡 Python offre anche `|` (unione, tutti gli elementi dei due insiemi) e `&` (intersezione, solo gli elementi presenti in entrambi). Qui, a differenza dei booleani, `|` e `&` **sono** il modo corretto di usarli con gli insiemi.

### 2.4 Metodi

| Metodo | Cosa fa |
| --- | --- |
| `.add(valore)` | Aggiunge un elemento all'insieme |
| `.remove(valore)` | Rimuove un elemento dall'insieme |

```python
corsi_disponibili = {'Scientifico', 'Umanistico', 'Biologico'}
print(corsi_disponibili)
```

```python
# .add(): aggiunge un elemento (se esiste già, l'insieme non cambia)
corsi_disponibili.add('Sanitario')
print(corsi_disponibili)
```

```python
# .remove(): rimuove un elemento esistente
corsi_disponibili.remove('Sanitario')
print(corsi_disponibili)
```

### 2.5 Conversione

È molto comune convertire una **lista in insieme** (per eliminare le ripetizioni) e, poi, l'**insieme di nuovo in lista** (per poter usare di nuovo indici e metodi delle liste).

```python
squadre_paulistane = ['São Paulo', 'Palmeiras', 'Corinthians', 'Santos']

# list -> set: elimina eventuali ripetizioni
squadre_senza_ripetizione = set(squadre_paulistane)
print(squadre_senza_ripetizione)
print(type(squadre_senza_ripetizione))
```

```python
# set -> list: per tornare a usare indici e metodi delle liste
squadre_lista = list(squadre_senza_ripetizione)
print(squadre_lista)
print(type(squadre_lista))
```

### 2.6 Rivisitando la motivazione

Combinando le due conversioni, risolviamo il problema degli hashtag in un'unica riga: uniamo le liste, convertiamo in `set` (rimuovendo le ripetizioni) e torniamo a `list`.

```python
hashtag_settimana_unici = list(set(hashtag_lunedi + hashtag_martedi + hashtag_mercoledi + hashtag_giovedi + hashtag_venerdi))

print(hashtag_settimana_unici)
print(f'Totale hashtag unici: {len(hashtag_settimana_unici)}')
```

## 3. Dizionari

Liste e insiemi conservano valori "sciolti", identificati solo dalla posizione o dalla loro stessa esistenza. Ma molti dati del mondo reale arrivano in **coppie**: un nome di rete e la sua password, un codice fiscale e il nome del cliente. Per questo usiamo i **dizionari**.

### 3.1 Motivazione

Per connetterti a una rete wi-fi, hai bisogno di due informazioni: il **nome della rete** e la **password di accesso**. Cercando le reti disponibili, il tuo dispositivo trova una lista di nomi:

```python
reti_wifi_disponibili = ['rete1', 'cnx_cnx', 'uai-fi', 'r3d3']
print(reti_wifi_disponibili)
```

Il problema è che questa lista salva solo i **nomi**. Non c'è modo di associare ogni rete alla sua password corrispondente. **Come memorizziamo coppie di informazioni collegate, come nome e password?**

### 3.2 Definizione

I **dizionari** memorizzano dati nel formato **chiave-valore**: ogni elemento è composto da una **chiave** (usata per cercare il dato) e un **valore** (il dato stesso). Sono del tipo `dict` e vengono creati tra parentesi graffe `{}`, nel formato `{chiave: valore}`.

```python
brasile = {'capitale': 'Brasília', 'lingua': 'Portoghese', 'popolazione_milioni': 210}

print(brasile)
print(type(brasile))
```

> ⚠️ **Attenzione:** non è permesso avere **chiavi duplicate**. Se ripeti una chiave, l'ultimo valore assegnato a essa **sovrascrive** i precedenti, senza generare errore:

```python
auto = {
    'marca': 'Volkswagen',
    'modello': 'Polo',
    'anno': 2021,
    'anno': 2004,  # sovrascrive il valore precedente della chiave 'anno'
}

print(auto)
```

Il valore di un dizionario può essere anche **un altro dizionario** — questo si chiama **dizionario composto**, molto utile per rappresentare dati con una gerarchia:

```python
anagrafica = {
    'andre': {
        'nome': 'André Perez',
        'anno_nascita': 1992,
        'padre': {'nome': 'Carlos Perez', 'anno_nascita': 1965},
        'madre': {'nome': 'Marta Perez', 'anno_nascita': 1968},
    }
}

print(anagrafica)
```

Per accedere a un valore attraverso più livelli, concateniamo le parentesi quadre con ogni chiave, dal livello più esterno al più interno:

```python
anno_nascita_madre = anagrafica['andre']['madre']['anno_nascita']
print(anno_nascita_madre)
```

### 3.3 Operazioni

A differenza delle liste, gli elementi di un dizionario si accedono e si aggiornano tramite la **chiave**, non tramite la posizione.

```python
punteggio_credito = {'123': 750, '789': 980}
```

```python
# Accesso a un valore tramite la chiave
punteggio_cliente_123 = punteggio_credito['123']
punteggio_cliente_789 = punteggio_credito['789']

print(punteggio_cliente_123)
print(punteggio_cliente_789)
```

```python
# Aggiornamento di un valore: assegniamo un nuovo valore a una chiave già esistente
punteggio_credito['123'] = 435
print(punteggio_credito)
```

```python
# Aggiunta di una nuova coppia chiave-valore: basta assegnare a una chiave nuova
punteggio_credito['456'] = 1000
print(punteggio_credito)
```

> ⚠️ **Attenzione:** accedere a una chiave che **non esiste** con `dizionario[chiave]` genera un errore `KeyError`. Per cercare in modo sicuro, usa il metodo `.get(chiave)`, che restituisce `None` (invece di un errore) quando la chiave non esiste.

```python
print(punteggio_credito.get('456'))  # esiste: restituisce il valore
print(punteggio_credito.get('999'))  # non esiste: restituisce None, senza errore
```

### 3.4 Metodi

| Metodo | Cosa fa |
| --- | --- |
| `.update(dict)` | Aggiunge o aggiorna più coppie chiave-valore in una volta |
| `.pop(chiave)` | Rimuove un elemento tramite la chiave e **restituisce** il valore rimosso |
| `.get(chiave)` | Restituisce il valore della chiave, o `None` se non esiste |

```python
articolo = dict(
    titolo='Modulo 02 - Strutture di Dati',
    corpo='Argomenti, Lezioni, Liste, Insiemi, Dizionari...',
    totale_caratteri=1530,
)
print(articolo)
```

```python
# .update(): aggiorna una o più coppie in una volta
articolo.update({'totale_caratteri': 7850})
print(articolo)
```

```python
# .pop(): rimuove per CHIAVE e restituisce il valore rimosso
totale_caratteri = articolo.pop('totale_caratteri')

print(f'Valore rimosso: {totale_caratteri}')
print(articolo)
```

### 3.5 Conversione

Possiamo convertire le **chiavi** e i **valori** di un dizionario in liste, con i metodi `.keys()` e `.values()`:

```python
articolo = dict(
    titolo='Modulo 02 - Strutture di Dati',
    corpo='Argomenti, Lezioni, Liste, Insiemi, Dizionari...',
    totale_caratteri=1530,
)
```

```python
chiavi = list(articolo.keys())

print(chiavi)
print(type(chiavi))
```

```python
valori = list(articolo.values())

print(valori)
print(type(valori))
```

### 3.6 Rivisitando la motivazione

Ora possiamo rappresentare ogni rete wi-fi come un dizionario con le chiavi `nome` e `password`, e salvarle tutte in una lista:

```python
reti_wifi_disponibili = []

rete = {'nome': 'rete1', 'password': 'cnx_cnx'}
reti_wifi_disponibili.append(rete)

rete = {'nome': 'uai-fi', 'password': 'r3d3'}
reti_wifi_disponibili.append(rete)

print(reti_wifi_disponibili)
```

Ora sì: per ogni rete della lista, sappiamo esattamente qual è il nome e qual è la password corrispondente — il problema iniziale è stato risolto combinando **lista** (la collezione di reti) e **dizionario** (i dati di ogni rete).

## Riepilogo del Modulo

| Struttura | Tipo | Ordinata? | Ripete valori? | Accesso | Metodi principali |
| --- | --- | --- | --- | --- | --- |
| Lista | `list` | Sì | Sì | Per indice `lista[i]` | `.append()`, `.insert()`, `.remove()`, `.pop()` |
| Insieme | `set` | No | No | Non indicizzato | `.add()`, `.remove()`, `-`, `\|`, `&` |
| Dizionario | `dict` | Sì (dal Python 3.7) | Le chiavi no, i valori sì | Per chiave `dic[chiave]` | `.update()`, `.pop()`, `.get()`, `.keys()`, `.values()` |

Funzioni native viste in questo modulo: `list()`, `set()`, `dict()`, `sum()`, `len()`.
