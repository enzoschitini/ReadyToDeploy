## 1. Struttura condizionale if / elif / else

La maggior parte dei programmi deve comportarsi in modo diverso a seconda della situazione: approvare o rifiutare un pagamento, concedere o bloccare un accesso, applicare uno sconto oppure no. Questo comportamento è controllato dal **flusso condizionale**, e in Python viene scritto con le parole chiave `if`, `elif` e `else`.

### 1.1 Motivazione

Stai sviluppando il sistema di pagamento con carta di credito di un *e-commerce*. Prima di approvare l'acquisto, il sistema deve confrontare il codice di sicurezza (CVV) fornito dal cliente con il codice registrato nella banca dati.

````python
# Codice digitato dal cliente e codice salvato nell'anagrafica
codigo_de_seguranca_informado = '291'
codigo_de_seguranca_cadastro = '010'
````

**Come facciamo eseguire al programma un pezzo di codice diverso a seconda che i due codici siano uguali o no?**

### 1.2 if / else

La struttura `if` / `else` esegue un blocco di codice **se** una condizione è `True`, e un altro blocco **altrimenti** (quando è `False`):

```python
if condizione:
  # eseguito quando condizione è True
else:
  # eseguito quando condizione è False
```

> ⚠️ **Attenzione:** Python usa l'**indentazione** (il rientro all'inizio della riga) per sapere cosa appartiene a ciascun blocco. Tutte le righe di uno stesso blocco devono avere lo stesso rientro, altrimenti il codice non funziona.

````python
esta_chovendo = False

if esta_chovendo:
  print('Porta un ombrello')
else:
  print('Non serve l\'ombrello')
````

Applicando al nostro problema: confrontiamo i due codici con `==` (visto nel Modulo 01) e usiamo il risultato come condizione.

````python
if codigo_de_seguranca_informado == codigo_de_seguranca_cadastro:
  print('Pagamento approvato')
else:
  print('Errore: codice di sicurezza non valido')
````

### 1.3 Condizioni composte con and / or

Molte volte è necessario valutare **più di una condizione** contemporaneamente. Per farlo, combiniamo gli operatori logici `and` e `or` (visti nel Modulo 01) direttamente nella condizione dell'`if`.

**Esempio:** oltre al codice di sicurezza, il sistema richiede anche che la password inserita sia corretta. Il pagamento viene approvato solo se **entrambe** le condizioni sono vere.

````python
codigo_de_seguranca_informado = '852'
codigo_de_seguranca_cadastro = '852'

senha_informada = '7783'
senha_cadastro = '7783'
````

````python
# Le parentesi in ogni confronto rendono la condizione più facile da leggere
if (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Pagamento approvato')
else:
  print('Errore: pagamento non autorizzato')
````

> 💡 Puoi trovare anche gli operatori `&` (e) e `|` (o) in librerie come **Pandas**. Nel Python "puro", però, dentro un `if`, la forma consigliata e più leggibile è usare `and` e `or`.

### 1.4 if / elif / else

Quando esistono **più di due** possibilità, usiamo `elif` (abbreviazione di "*else if*") per concatenare diverse condizioni. Python testa ogni condizione **nell'ordine in cui appare** ed esegue il blocco della **prima** che risulta vera — ignorando tutte le altre, anche se anch'esse sono vere.

```python
if condizione_1:
  # eseguito se condizione_1 è True
elif condizione_2:
  # eseguito se condizione_1 è False e condizione_2 è True
else:
  # eseguito se nessuna delle precedenti è True
```

### 1.5 Rivedendo la motivazione

Ora abbiamo tutto per gestire le **quattro** combinazioni possibili tra codice di sicurezza e password, ciascuna con un messaggio diverso:

| Codice corretto? | Password corretta? | Risultato |
| --- | --- | --- |
| Sì | Sì | Pagamento approvato |
| No | Sì | Errore: codice di sicurezza non valido |
| Sì | No | Errore: password non valida |
| No | No | Errore: codice di sicurezza e password non validi |

````python
codigo_de_seguranca_informado = '802'
codigo_de_seguranca_cadastro = '852'

senha_informada = '7703'
senha_cadastro = '7783'
````

````python
if (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Pagamento approvato')
elif (codigo_de_seguranca_informado != codigo_de_seguranca_cadastro) and (senha_informada == senha_cadastro):
  print('Errore: codice di sicurezza non valido')
elif (codigo_de_seguranca_informado == codigo_de_seguranca_cadastro) and (senha_informada != senha_cadastro):
  print('Errore: password non valida')
else:
  print('Errore: codice di sicurezza e password non validi')
````

> 💡 Prova a cambiare i valori di `codigo_de_seguranca_informado` e `senha_informada` sopra ed esegui di nuovo le due celle, per vedere apparire gli altri tre messaggi.

## 2. Gestione delle eccezioni con try / except / finally

Anche un codice senza errori di battitura può fallire **durante l'esecuzione**, a causa di situazioni impreviste: una divisione per zero, un dato nel formato sbagliato, una posizione che non esiste in una lista. Questi errori di esecuzione sono chiamati **eccezioni**, e Python permette di gestirle in modo che il programma non venga interrotto.

### 2.1 Motivazione

Stai calcolando quanto deve pagare ciascuna persona in un conto di ristorante, dividendo l'importo totale per il numero di persone. Solo che questo numero di persone arriva da un modulo compilato da qualcun altro, e a volte arriva sbagliato — persino come zero.

Se eseguiamo la divisione direttamente, Python interrompe il programma:

```python
preco_total_conta = 132.85
quantidade_de_pessoas = 0

valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
```

```
ZeroDivisionError: float division by zero
```

**Come evitiamo che il programma si blocchi quando questo accade?**

### 2.2 Tipi comuni di eccezione

Ogni tipo di errore ha un nome proprio, il che aiuta a gestirlo in modo specifico:

| Eccezione | Si verifica quando... | Esempio |
| --- | --- | --- |
| `ZeroDivisionError` | una divisione ha divisore uguale a zero | `10 / 0` |
| `TypeError` | un'operazione combina tipi incompatibili | `'Età: ' + 19` |
| `ValueError` | un valore non può essere convertito nel tipo richiesto | `int('abc')` |
| `IndexError` | una posizione non esiste in una lista | `lista[10]` (lista con 3 elementi) |
| `KeyError` | una chiave non esiste in un dizionario | `dicionario['chiave_inesistente']` |
| `NameError` | una variabile usata non è stata definita | `print(variabile_che_non_esiste)` |

### 2.3 try / except

La struttura `try` / `except` esegue un blocco di codice e, **se** si verifica un'eccezione, esegue un altro blocco al suo posto — senza interrompere il programma.

```python
try:
  # codice che può fallire
except TipoDiErrore:
  # eseguito solo se questa eccezione si verifica
```

````python
preco_total_conta = 132.85
quantidade_de_pessoas = 0

try:
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Ogni persona paga R$ {valor_por_pessoa:.2f}')
except ZeroDivisionError:
  print('Errore: il numero di persone deve essere maggiore di zero.')
````

Con un numero di persone valido, il `try` funziona normalmente e l'`except` non viene nemmeno valutato:

````python
preco_total_conta = 132.85
quantidade_de_pessoas = 2

try:
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Ogni persona paga R$ {valor_por_pessoa:.2f}')
except ZeroDivisionError:
  print('Errore: il numero di persone deve essere maggiore di zero.')
````

### 2.4 Catturare i dettagli dell'eccezione e gestire più tipi

Possiamo catturare l'eccezione in una variabile (convenzione: `except ... as exc`) per esaminarne il messaggio, e possiamo anche gestire **diversi tipi di eccezione** nello stesso `try`, ciascuno con il proprio `except`.

**Esempio:** cerchiamo una posizione che non esiste in una lista di anni.

````python
anos_com_vendas = [2019, 2020, 2021]

try:
  ano_mais_recente = anos_com_vendas[5]
  print(ano_mais_recente)
except IndexError as exc:
  print(f'Errore: posizione non valida. La lista ha solo {len(anos_com_vendas)} anni.')
  print(f'Messaggio originale di Python: {exc}')
````

**Esempio:** convertiamo un testo proveniente da un modulo in numero, gestendo prima l'errore specifico e per ultimo un errore generico.

> ⚠️ **Attenzione:** l'ordine è importante. Python testa i blocchi `except` nell'ordine in cui appaiono, quindi metti sempre gli errori **più specifici per primi** e un `except Exception` **per ultimo**, come rete di sicurezza.

````python
dado_de_entrada = 'non è un numero'

try:
  valor_convertido = int(dado_de_entrada)
  print(valor_convertido)
except ValueError:
  print('Errore: non è stato possibile convertire il testo in numero.')
except Exception as exc:
  print(f'Errore inaspettato: {exc}')
````

### 2.5 try / except / finally

Il blocco `finally` viene eseguito **sempre**, che si sia verificata un'eccezione oppure no. È utile per azioni che devono avvenire indipendentemente dal risultato, come confermare che un processo è terminato.

````python
nome_cliente = 'André'
idade_cliente = 19

try:
  # idade_cliente è int; il + non unisce str con int (TypeError)
  apresentacao = 'Cliente: ' + nome_cliente + ', età: ' + idade_cliente
except TypeError:
  apresentacao = 'Cliente: ' + nome_cliente + ', età: ' + str(idade_cliente)
finally:
  print(apresentacao)
  print('Registrazione elaborata.')
````

### 2.6 Rivedendo la motivazione

Rendiamo il calcolo del valore a persona più realistico: il numero di persone arriva come **testo** dal modulo, quindi può fallire in due modi diversi — un testo che non è un numero (`ValueError`) oppure il numero zero (`ZeroDivisionError`). Gestiamo entrambi, e usiamo il `finally` per avvisare che il calcolo è terminato.

````python
preco_total_conta = 132.85
quantidade_de_pessoas_texto = '0'  # arrivato come testo da un modulo

try:
  quantidade_de_pessoas = int(quantidade_de_pessoas_texto)
  valor_por_pessoa = preco_total_conta / quantidade_de_pessoas
  print(f'Ogni persona paga R$ {valor_por_pessoa:.2f}')
except ValueError:
  print('Errore: il numero di persone deve essere un numero.')
except ZeroDivisionError:
  print('Errore: il numero di persone deve essere maggiore di zero.')
finally:
  print('Calcolo completato.')
````

## 3. Struttura di ripetizione for

Buona parte del lavoro di un programma è ripetitiva: sommare tutti i valori di una lista, verificare ogni riga di un foglio di calcolo, elaborare ogni cliente di un'anagrafica. Invece di ripetere il codice manualmente, usiamo la struttura di ripetizione `for`.

### 3.1 Motivazione

Hai le vendite del giorno di un negozio e vuoi sapere due cose: il **totale venduto** e **quante vendite** hanno superato R\$ 300,00.

````python
vendas_do_dia = [361, 553, 194, 13, 510, 33, 135]
````

**Come elaboriamo tutti questi valori senza scrivere una riga di codice per ciascuno?**

### 3.2 for / range

La struttura `for` esegue un blocco di codice **una volta per ogni elemento** di una collezione. Combinata con la funzione nativa `range()`, ripete il blocco un numero determinato di volte:

```python
for variabile_temporanea in range(inizio, fine, passo):
  # eseguito una volta per ogni valore nell'intervallo
```

````python
for valor in range(6):
  print(valor)
````

> 💡 `range(fine)` inizia da 0; `range(inizio, fine)` definisce l'inizio; `range(inizio, fine, passo)` definisce anche l'incremento. Come nello slicing delle *stringhe* (Modulo 01), il valore `fine` **non** viene mai incluso.

````python
# Multipli di 3, da 0 (incluso) a 15 (escluso)
for multiplo_de_tres in range(0, 15, 3):
  print(multiplo_de_tres)
````

### 3.3 for / liste

Il modo più comune di usare `for` è per scorrere gli elementi di una lista, uno alla volta:

````python
for venda in vendas_do_dia:
  print(venda)
````

Uno schema molto comune è **accumulare** un risultato a ogni ripetizione, usando una variabile creata **prima** del `for`:

````python
total_vendido = 0

for venda in vendas_do_dia:
  total_vendido += venda

print(f'Totale venduto: R$ {total_vendido:.2f}')
````

E possiamo combinare `for` con `if` (visto nella sezione 1) per contare solo gli elementi che soddisfano una condizione:

````python
quantidade_vendas_acima_300 = 0

for venda in vendas_do_dia:
  if venda > 300:
    quantidade_vendas_acima_300 += 1

print(f'Vendite sopra R$ 300: {quantidade_vendas_acima_300}')
````

> 💡 Anche le *stringhe* sono sequenze e possono essere scorse carattere per carattere. L'operatore `in` qui verifica se un carattere **appartiene** a un altro testo.

````python
frase = 'Ready To Deploy'
quantidade_vogais = 0

for caractere in frase.lower():
  if caractere in 'aeiou':
    quantidade_vogais += 1

print(f'Numero di vocali: {quantidade_vogais}')
````

### 3.4 for / dizionari

I dizionari hanno tre metodi per essere scorsi in un `for`: `.items()` (chiave e valore), `.keys()` (solo le chiavi) e `.values()` (solo i valori).

````python
escore_de_credito = {'123.456.789-00': 750, '987.654.321-00': 812, '111.222.333-44': 980}
````

````python
# .items(): restituisce ogni coppia chiave-valore
for documento, escore in escore_de_credito.items():
  print(f'Documento {documento}: punteggio {escore}')
````

````python
# .keys(): restituisce solo le chiavi
for documento in escore_de_credito.keys():
  print(documento)
````

````python
# .values(): restituisce solo i valori
for escore in escore_de_credito.values():
  print(escore)
````

> 💡 `for documento in escore_de_credito:` (senza `.keys()`) ha esattamente lo stesso effetto, perché scorrere direttamente un dizionario restituisce già le sue chiavi.

### 3.5 break / continue

Due parole chiave controllano l'andamento di un ciclo `for`:

| Parola chiave | Effetto |
| --- | --- |
| `break` | interrompe il ciclo **immediatamente**, senza eseguire le ripetizioni rimanenti |
| `continue` | salta direttamente alla **ripetizione successiva**, senza eseguire il resto del blocco corrente |

**Esempio:** fermarsi non appena troviamo la prima vendita sopra R\$ 500,00.

````python
for venda in vendas_do_dia:
  if venda > 500:
    print(f'Prima vendita sopra R$ 500: {venda}')
    break
````

**Esempio:** mostrare solo le vendite con valore **pari**, saltando quelle dispari con `continue`.

````python
for venda in vendas_do_dia:
  if venda % 2 != 0:
    continue
  print(f'Vendita pari: {venda}')
````

> ⚠️ **Attenzione:** qualsiasi codice scritto **dopo** un `break` o `continue`, ancora dentro lo stesso blocco `if`, non viene mai eseguito — il ciclo è già stato interrotto o avanzato prima di arrivarci.

### 3.6 Rivedendo la motivazione

Con `for`, `if` e l'accumulo in variabili, risolviamo le due domande della motivazione in un unico ciclo:

````python
vendas_do_dia = [361, 553, 194, 13, 510, 33, 135]

total_vendido = 0
quantidade_vendas_acima_300 = 0

for venda in vendas_do_dia:
  total_vendido += venda
  if venda > 300:
    quantidade_vendas_acima_300 += 1

print(f'Totale venduto: R$ {total_vendido:.2f}')
print(f'Vendite sopra R$ 300: {quantidade_vendas_acima_300}')
````
