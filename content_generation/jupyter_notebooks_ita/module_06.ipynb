## 1. Un po' di teoria

Prima di scrivere codice, vale la pena capire l'idea alla base della programmazione orientata agli oggetti. In questa sezione vedremo dove si colloca la POO tra gli stili di programmazione e conosceremo i tre concetti che sostengono il resto del modulo: **classe**, **oggetto** ed **ereditarietà**.


### 1.1 Paradigmi di programmazione

Un **paradigma di programmazione** è uno **stile** di organizzare e scrivere codice. I tre più conosciuti sono:

| Paradigma | Idea centrale | Domanda a cui il codice risponde |
| --- | --- | --- |
| **Imperativo** | Una sequenza di istruzioni che modificano variabili passo dopo passo | "Come fare?" |
| **Funzionale** | Composizione di funzioni che ricevono valori e restituiscono risultati, senza alterare dati esterni | "Cosa calcolare?" |
| **Orientato agli oggetti** | Dati e comportamenti raggruppati in **oggetti**, che rappresentano elementi del problema | "Chi fa cosa?" |

Python è un linguaggio **multiparadigma**: permette di usare i tre stili, anche mescolati nello stesso programma. Per confrontarli, vediamo lo stesso problema risolto nei tre stili: calcolare il totale di un carrello della spesa.


**Stile imperativo:** scorriamo i prezzi e accumuliamo il totale in una variabile.


````python
precos_carrinho = [49.90, 15.00, 120.50]

# Passo dopo passo: la variabile total_imperativo viene modificata a ogni giro del ciclo
total_imperativo = 0
for preco in precos_carrinho:
    total_imperativo += preco

print(f'Totale (imperativo): {total_imperativo:.2f}')
````

**Stile funzionale:** descriviamo il risultato combinando funzioni, senza modificare nessuna variabile lungo il percorso.


````python
# sum() riceve la lista e restituisce il totale, senza modificare nulla
total_funcional = sum(precos_carrinho)

print(f'Totale (funzionale): {total_funcional:.2f}')
````

**Stile orientato agli oggetti:** creiamo un "carrello" che conserva i propri articoli e sa calcolare da solo il proprio totale. Non preoccuparti della sintassi adesso: è esattamente ciò che impareremo in questo modulo.


````python
class Carrinho:

    def __init__(self):
        self.precos = []

    def adicionar(self, preco: float) -> None:
        self.precos.append(preco)

    def total(self) -> float:
        return sum(self.precos)


# Il carrello conserva i dati E le azioni che hanno senso per lui
meu_carrinho = Carrinho()
meu_carrinho.adicionar(49.90)
meu_carrinho.adicionar(15.00)
meu_carrinho.adicionar(120.50)

print(f'Totale (orientato agli oggetti): {meu_carrinho.total():.2f}')
````

> 💡 **Suggerimento:** nessun paradigma è "migliore" in tutti i casi. Per un calcolo semplice, una riga come `sum(precos_carrinho)` è sufficiente. La programmazione orientata agli oggetti brilla quando il sistema cresce e inizia ad avere molti elementi (clienti, prodotti, ordini), ciascuno con i propri dati e le proprie regole.


### 1.2 Classe

Una **classe** è la rappresentazione di un elemento, reale o astratto, e funziona come una **ricetta** (o uno stampo) per creare oggetti. Essa definisce:

- quali **caratteristiche** ha ogni elemento di quel tipo (gli **attributi**);
- quali **azioni** può compiere (i **metodi**).

**Esempio:** la classe `Pessoa` definisce che ogni persona ha nome, età e documento, e che ogni persona può parlare e dormire.


### 1.3 Oggetto

Un **oggetto** è un'**istanza** di una classe, cioè qualcosa di concreto creato a partire dalla ricetta. La classe descrive; l'oggetto esiste e ha valori propri.

**Esempio:** André, 30 anni, è un'istanza della classe `Pessoa`. Maria, 56 anni, è un'altra istanza della stessa classe. Entrambe seguono la stessa ricetta, ma ciascuna ha i propri dati.

Pensa alla planimetria di una casa: la planimetria è la **classe**, e ogni casa costruita a partire da essa è un **oggetto**. Con un'unica planimetria, possiamo costruire quante case vogliamo, e ognuna può avere un colore diverso.


### 1.4 Ereditarietà

L'**ereditarietà** è un modo per **specializzare** una classe: creiamo una nuova classe che riutilizza tutto ciò che un'altra classe già possiede e aggiunge ciò che le è specifico.

**Esempio:** uno studente è un tipo speciale di persona. Ha nome, età e documento (come qualsiasi persona), ma studia anche in un'università.

| Concetto | Analogia | Esempio |
| --- | --- | --- |
| Classe | La ricetta o la planimetria | `Pessoa` |
| Oggetto | Il piatto pronto o la casa costruita | André, Maria, Pedro |
| Ereditarietà | Una variante della ricetta | `Estudante` è una `Pessoa` |

## 2. Classi

Le classi sono il punto di partenza della programmazione orientata agli oggetti: è in esse che definiamo quali dati e quali comportamenti avrà un tipo di elemento. In questa sezione costruiremo, passo dopo passo, la classe `Pessoa`.


### 2.1 Motivazione

Lavori nel sistema di registrazione di una banca e devi conservare i dati di diversi clienti. Con quanto visto finora, un'opzione potrebbe essere usare variabili separate:


````python
# Ogni informazione di ogni cliente in una variabile diversa
nome_cliente_1 = 'Andre Perez'
idade_cliente_1 = 30

nome_cliente_2 = 'Maria Perez'
idade_cliente_2 = 56

print(f'{nome_cliente_1}, {idade_cliente_1} anni')
print(f'{nome_cliente_2}, {idade_cliente_2} anni')
````

Con 2 clienti funziona. Ma con 2 mila? E quando ogni cliente avrà 15 informazioni? E se ogni regola del sistema (fare gli auguri di compleanno, validare il documento, mostrare i dati) dovesse ricevere tutte queste variabili come parametro? Il codice diventa ripetitivo e facile da rompere.

**Come possiamo raggruppare i dati di un cliente e le azioni che hanno senso per lui in un'unica struttura?**


### 2.2 Definizione

Una **classe** è una ricetta per la creazione di oggetti. La struttura generale è:

```python
class NomeDaClasse:

    def __init__(self, parametros):
        self.atributo_1 = ...
        self.atributo_2 = ...

    def metodo_1(self, parametros):
        ...

    def metodo_2(self, parametros):
        ...
```

Alcuni punti importanti di questa struttura:

- La parola riservata `class` avvia la definizione della classe;
- Per convenzione ([PEP 8](https://peps.python.org/pep-0008/#class-names)), i nomi delle classi usano il **`PascalCase`**: ogni parola inizia con la lettera maiuscola e non ci sono `_` (es: `Pessoa`, `ContaBancaria`, `ArquivoCSV`). Variabili e funzioni continuano a usare lo `snake_case`;
- Tutto ciò che appartiene alla classe viene **indentato** al suo interno (4 spazi).


Creiamo la versione più semplice possibile della classe `Pessoa`. La parola `pass` indica un blocco vuoto: la classe esiste, ma non fa ancora nulla.


````python
class Pessoa:
    pass


print(Pessoa)
````

> 💡 **Suggerimento:** nel codice più datato troverai `class Pessoa(object):`. In Python 3, ogni classe eredita già automaticamente da `object` (vedremo l'ereditarietà nella sezione 4), quindi `class Pessoa:` è equivalente ed è la forma consigliata oggi.


### 2.3 Attributi

Gli **attributi** rappresentano le **caratteristiche** della classe. Vengono creati all'interno di un metodo speciale chiamato **`__init__`** (da *initialize*, "inizializzare"), che Python esegue automaticamente ogni volta che viene creato un nuovo oggetto. È conosciuto come il **costruttore** della classe.

Due dettagli che spesso confondono all'inizio:

| Elemento | Cosa è |
| --- | --- |
| `self` | Un riferimento all'**oggetto stesso** che si sta creando. Ogni metodo della classe riceve `self` come primo parametro |
| `self.nome = nome` | Crea l'attributo `nome` **nell'oggetto** e vi salva il valore ricevuto nel parametro `nome` |


````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str):
        # Ogni self.atributo viene salvato dentro l'oggetto
        self.nome = nome
        self.idade = idade
        self.documento = documento
````

> 💡 **Suggerimento:** le annotazioni `: str` e `: int` sono **suggerimenti di tipo** (*type hints*). Documentano che tipo di valore ci si aspetta per ogni parametro e aiutano l'editor a segnalare errori, ma Python **non le impone**: sono solo documentazione.


Come nelle funzioni, i parametri di `__init__` possono avere **valori predefiniti**. Non ogni persona fornisce il documento in fase di registrazione, quindi lo rendiamo opzionale, con valore `None`:


````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento  # None quando non viene fornito
````

### 2.4 Metodi

I **metodi** rappresentano le **azioni** della classe. Sono funzioni definite **dentro** la classe e, per questo, ricevono sempre `self` come primo parametro. È tramite `self` che un metodo accede (e modifica) gli attributi dell'oggetto stesso.

In questa versione, `Pessoa` guadagna tre azioni:

| Metodo | Cosa fa |
| --- | --- |
| `dormir(horas)` | Mostra il progresso del sonno, ora per ora |
| `falar(texto)` | Mostra una frase pronunciata dalla persona |
| `fazer_aniversario()` | **Modifica** l'attributo `idade`, aggiungendo 1 |


````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento

    def dormir(self, horas: int) -> None:
        # self.nome: il metodo accede a un attributo dell'oggetto stesso
        for hora in range(1, horas + 1):
            print(f'{self.nome} sta dormendo... {hora}h di {horas}h')

    def falar(self, texto: str) -> None:
        print(f'{self.nome} dice: {texto}')

    def fazer_aniversario(self) -> None:
        # Un metodo può anche MODIFICARE lo stato dell'oggetto
        self.idade += 1
        print(f'Auguri, {self.nome}! Ora hai {self.idade} anni.')
````

> 💡 **Suggerimento:** l'annotazione `-> None` indica che il metodo **non restituisce** alcun valore (mostra solo qualcosa o modifica l'oggetto). Un metodo che restituisce un testo sarebbe annotato con `-> str`.


### 2.5 Il metodo speciale `__str__`

I metodi il cui nome inizia e termina con due trattini bassi (come `__init__`) sono **metodi speciali** (chiamati anche *dunder methods*, da *double underscore*). Python li richiama automaticamente in situazioni specifiche.

Uno dei più utili è **`__str__`**, richiamato quando l'oggetto viene convertito in testo, per esempio dentro un `print()`. Vediamo cosa succede quando stampiamo un oggetto della classe attuale, che **non ha ancora** `__str__` (la creazione di oggetti sarà approfondita nella sezione 3):


````python
# Senza __str__, Python mostra solo la classe e l'indirizzo di memoria dell'oggetto
pessoa_teste = Pessoa(nome='Andre Perez', idade=30)
print(pessoa_teste)
````

Poco utile, vero? Aggiungiamo `__str__` alla classe. Deve **restituire** (e non stampare) una *stringa* che descriva l'oggetto. Questa è la versione finale della classe `Pessoa`, che useremo nel resto del modulo:


````python
class Pessoa:

    def __init__(self, nome: str, idade: int, documento: str = None):
        self.nome = nome
        self.idade = idade
        self.documento = documento

    def dormir(self, horas: int) -> None:
        for hora in range(1, horas + 1):
            print(f'{self.nome} sta dormendo... {hora}h di {horas}h')

    def falar(self, texto: str) -> None:
        print(f'{self.nome} dice: {texto}')

    def fazer_aniversario(self) -> None:
        self.idade += 1
        print(f'Auguri, {self.nome}! Ora hai {self.idade} anni.')

    def __str__(self) -> str:
        # Evita di mostrare "None" quando il documento non è stato fornito
        if self.documento is None:
            documento_exibido = 'non fornito'
        else:
            documento_exibido = self.documento
        return f'{self.nome}, {self.idade} anni, documento: {documento_exibido}'
````

````python
# Ora print() usa __str__ e mostra una descrizione leggibile
pessoa_teste = Pessoa(nome='Andre Perez', idade=30)
print(pessoa_teste)
````

> ⚠️ **Attenzione:** ogni volta che modifichi una classe, è necessario **eseguire di nuovo** la cella che la definisce **e** ricreare gli oggetti. Gli oggetti creati prima continuano a seguire la versione precedente della classe.

## 3. Oggetti

Una classe da sola è solo una ricetta: sono gli **oggetti** a conservare davvero i dati e a eseguire le azioni. In questa sezione creeremo oggetti, accederemo e modificheremo i loro attributi, chiameremo i loro metodi e applicheremo tutto ciò in un esempio reale di lettura di file.


### 3.1 Definizione

Un **oggetto** è un'**istanza** di una classe. Per crearlo, "chiamiamo" la classe come se fosse una funzione, passando i parametri di `__init__` (senza `self`, che Python compila da solo). Poi usiamo il **punto** (`.`) per accedere ad attributi e metodi:

```python
objeto = NomeDaClasse(parametros)

objeto.atributo    # accede a una caratteristica
objeto.metodo()    # esegue un'azione
```


````python
# Tre oggetti creati dalla stessa classe, ciascuno con i propri dati
andre = Pessoa(nome='Andre Perez', idade=30, documento='123')
maria = Pessoa(nome='Maria Perez', idade=56, documento='456')
pedro = Pessoa(nome='Pedro Perez', idade=8)  # senza documento: usa il valore predefinito None

print(andre)
print(maria)
print(pedro)
````

Possiamo confermare a quale classe appartiene un oggetto con `type()` oppure, in modo più comune nella pratica, con `isinstance()`, che risponde con un booleano:


````python
print(type(andre))
print(isinstance(andre, Pessoa))
print(isinstance('Andre Perez', Pessoa))  # una stringa non è una Pessoa
````

### 3.2 Manipolare gli attributi

Un attributo funziona come una variabile che "vive" dentro l'oggetto: possiamo **leggerne** il valore, usarlo in calcoli e funzioni e anche **modificarlo**.


````python
# Leggendo gli attributi
print(andre.nome)
print(andre.idade)
````

**Esempio:** usare un attributo come argomento di una funzione comune.


````python
def maior_de_idade(idade: int) -> bool:
    return idade >= 18


# Passiamo solo l'attributo idade di ogni oggetto alla funzione
for pessoa in [andre, maria, pedro]:
    if maior_de_idade(idade=pessoa.idade):
        print(f'{pessoa.nome} è maggiorenne')
    else:
        print(f'{pessoa.nome} è minorenne')
````

**Esempio:** consultare lo *score* di credito di un cliente a partire dal documento. Il documento dell'oggetto funge da chiave per cercare il valore in un dizionario:


````python
# Score di credito di ogni cliente, indicizzato per numero di documento
score_credito = {'123': 750, '456': 812, '789': 327}

score_andre = score_credito[andre.documento]
print(f'Score di {andre.nome}: {score_andre}')
````

> ⚠️ **Attenzione:** `pedro` è stato creato senza documento (`None`), quindi `score_credito[pedro.documento]` genererebbe un `KeyError`. Con il metodo `.get()` del dizionario, possiamo definire un valore predefinito per quando la chiave non esiste.


````python
# score_pedro = score_credito[pedro.documento]           # ❌ KeyError: None
score_pedro = score_credito.get(pedro.documento, 'nessuno score')  # ✅ valore predefinito

print(f'Score di {pedro.nome}: {score_pedro}')
````

Possiamo anche **modificare** un attributo direttamente, con un'assegnazione comune. La modifica vale solo per quell'oggetto:


````python
# Pedro ha portato il documento: aggiorniamo solo l'oggetto pedro
pedro.documento = '789'

print(pedro)
print(andre)  # andre non è stato influenzato
````

### 3.3 Chiamare metodi

I metodi vengono chiamati con la sintassi `objeto.metodo(argumentos)`. Nota che **non** passiamo `self`: Python capisce che `self` è l'oggetto che precede il punto.


````python
# self dentro dormir() è proprio andre
andre.dormir(horas=3)
````

````python
andre.falar(texto='Ciao a tutti!')
maria.falar(texto='Buongiorno!')
````

Il metodo `fazer_aniversario()` modifica lo stato dell'oggetto. Osserva l'età prima e dopo:


````python
print(maria)

maria.fazer_aniversario()

print(maria)
````

> ⚠️ **Attenzione:** per eseguire un metodo è obbligatorio usare le parentesi. Scrivere `maria.fazer_aniversario` (senza `()`) non genera errore, ma **non esegue** nulla: fa solo riferimento al metodo.


### 3.4 Rivisitando la motivazione

Ricordi la registrazione dei clienti della sezione 2.1? Con la classe `Pessoa`, ogni cliente diventa **un unico oggetto** che porta con sé i propri dati e le proprie azioni. Salvando gli oggetti in una lista, registrare 3 o 3 mila clienti richiede lo stesso codice:


````python
clientes = [
    Pessoa(nome='Andre Perez', idade=30, documento='123'),
    Pessoa(nome='Maria Perez', idade=56, documento='456'),
    Pessoa(nome='Pedro Perez', idade=8),
]

# Un'unica regola applicata a tutti i clienti
for cliente in clientes:
    print(cliente)
````

````python
# Cambio d'anno: tutti i clienti festeggiano il compleanno
for cliente in clientes:
    cliente.fazer_aniversario()
````

Nessuna variabile sparsa, nessuna funzione che riceve decine di parametri: ogni oggetto sa gestire i propri dati.


### 3.5 Tutto in Python è un oggetto

Usi oggetti fin dal primo modulo, forse senza accorgertene. Numeri, testi, liste, dizionari e persino funzioni sono oggetti di qualche classe nativa di Python. La funzione `type()` mostra la classe di ciascuno:


````python
tipos = [
    type(10),
    type(1.1),
    type('Ready To Deploy'),
    type(True),
    type(None),
    type([1, 2, 3]),
    type({1, 2, 3}),
    type({'janeiro': 1}),
    type(lambda x: x),
]

for tipo in tipos:
    print(tipo)
````

Nota il `<class '...'>` in ogni riga: `int`, `str`, `list` e gli altri sono **classi**. Per questo hanno metodi, che chiamiamo esattamente come abbiamo fatto con `Pessoa`:


````python
# nome è un oggetto della classe str, e upper() è un metodo di questa classe
nome = 'Andre Perez'
print(nome.upper())
````

````python
# juros è un oggetto della classe list, e pop() è un metodo di questa classe
juros = [0.02, 0.07, 0.15]

juro_removido = juros.pop(1)  # rimuove e restituisce l'elemento in posizione 1
print(juro_removido)
print(juros)
````

### 3.6 Ripasso: `strip()` e `split()`

Nel prossimo esempio leggeremo un file CSV riga per riga. Per farlo, due metodi delle *stringhe* saranno fondamentali:

| Metodo | Cosa fa | Esempio | Risultato |
| --- | --- | --- | --- |
| `.strip()` | Rimuove spazi e a capo (`\n`) dall'inizio e dalla fine | `' banana \n'.strip()` | `'banana'` |
| `.strip(caracteres)` | Rimuove i caratteri indicati dall'inizio e dalla fine | `'..banana..'.strip('.')` | `'banana'` |
| `.split(separador)` | Divide il testo in una **lista** a ogni separatore | `'a,b,c'.split(',')` | `['a', 'b', 'c']` |


````python
# strip() senza argomenti: rimuove spazi e a capo alle estremità
fruta = ' banana \n'
print(fruta.strip())
````

````python
# strip(caracteres): rimuove QUALSIASI dei caratteri indicati, solo alle estremità
fruta = ',,,,,rrttgg.....banana....rrr'
print(fruta.strip(',.grt'))
````

````python
# Combinando i due: prima puliamo le estremità, poi dividiamo
frase = ' O gato pulou sobre o muro '
palavras = frase.strip().split()  # senza argomento, split() divide sugli spazi
print(palavras)
````

````python
# Una tipica riga di file CSV: valori separati da virgola e un \n alla fine
linha_csv = '30,unemployed,married,primary\n'
valores = linha_csv.strip().split(',')
print(valores)
````

### 3.7 Esempio pratico: classe `ArquivoCSV`

I file **CSV** (*Comma-Separated Values*) sono tabelle in formato testo: ogni riga è un record e i valori sono separati da virgola. La prima riga di solito riporta il nome delle colonne.

Creiamo una classe che legge un CSV con dati di clienti di una banca e permette di estrarre qualsiasi colonna. Prima creiamo il file `banco.csv`:


````python
conteudo_csv = '''age,job,marital,education,default,balance,housing,loan
30,unemployed,married,primary,no,1787,no,no
33,services,married,secondary,no,4789,yes,yes
35,management,single,tertiary,no,1350,yes,no
30,management,married,tertiary,no,1476,yes,yes
59,blue-collar,married,secondary,no,0,yes,no
35,management,single,tertiary,no,747,no,no
36,self-employed,married,tertiary,no,307,yes,no
39,technician,married,secondary,no,147,yes,no
41,entrepreneur,married,tertiary,no,221,yes,no
43,services,married,primary,no,-88,yes,yes'''

# Salviamo il testo in un file nell'ambiente attuale
with open('banco.csv', mode='w', encoding='utf8') as arquivo:
    arquivo.write(conteudo_csv)
````

> 💡 **Suggerimento:** in Colab troverai anche il comando magico `%%writefile banco.csv` nella prima riga di una cella, che salva il contenuto della cella in un file. Qui usiamo `open()` perché funziona in qualsiasi ambiente Python.


Ora, la classe. Avrà due tipi di metodo:

| Metodo | Tipo | Cosa fa |
| --- | --- | --- |
| `_extrair_conteudo()` | Interno | Legge il file e restituisce la lista delle righe |
| `_extrair_nome_colunas()` | Interno | Restituisce i nomi delle colonne (la prima riga) |
| `extrair_coluna(indice_coluna)` | Pubblico | Restituisce i valori di una colonna a partire dalla sua posizione |
| `extrair_coluna_por_nome(nome_coluna)` | Pubblico | Restituisce i valori di una colonna a partire dal suo nome |

Il `_` all'inizio del nome è una **convenzione** di Python per indicare che un metodo è **a uso interno** della classe: esiste per aiutare gli altri metodi, non per essere chiamato dall'esterno.


````python
class ArquivoCSV:

    def __init__(self, caminho_arquivo: str):
        self.caminho_arquivo = caminho_arquivo
        # Il costruttore stesso legge già il file e scopre le colonne
        self.conteudo = self._extrair_conteudo()
        self.colunas = self._extrair_nome_colunas()

    def _extrair_conteudo(self) -> list:
        with open(self.caminho_arquivo, mode='r', encoding='utf8') as arquivo:
            return arquivo.readlines()

    def _extrair_nome_colunas(self) -> list:
        # La prima riga del file è l'intestazione
        return self.conteudo[0].strip().split(',')

    def extrair_coluna(self, indice_coluna: int) -> list:
        coluna = []
        # [1:] salta la prima riga, poiché l'intestazione non è un dato
        for linha in self.conteudo[1:]:
            valores_linha = linha.strip().split(',')
            coluna.append(valores_linha[indice_coluna])
        return coluna

    def extrair_coluna_por_nome(self, nome_coluna: str) -> list:
        # Un metodo può chiamare un altro metodo dello stesso oggetto tramite self
        indice_coluna = self.colunas.index(nome_coluna)
        return self.extrair_coluna(indice_coluna=indice_coluna)
````

Quando l'oggetto viene creato, `__init__` legge già il file e popola gli attributi `conteudo` e `colunas`:


````python
arquivo_banco = ArquivoCSV(caminho_arquivo='banco.csv')

print(arquivo_banco.colunas)
````

````python
# Ogni elemento dell'attributo conteudo è una riga del file (nota il \n alla fine)
print(arquivo_banco.conteudo[:3])
````

**Esempio:** estrarre la colonna `job` (professione), che si trova in posizione `1`.


````python
profissoes = arquivo_banco.extrair_coluna(indice_coluna=1)
print(profissoes)
````

**Esempio:** estrarre la colonna `education` (istruzione), che si trova in posizione `3`.


````python
escolaridades = arquivo_banco.extrair_coluna(indice_coluna=3)
print(escolaridades)
````

**Esempio:** memorizzare la posizione di ogni colonna è faticoso e facile da sbagliare. Con `extrair_coluna_por_nome()`, basta indicare il nome:


````python
estados_civis = arquivo_banco.extrair_coluna_por_nome(nome_coluna='marital')
print(estados_civis)

# Il risultato è una lista comune, quindi possiamo usare i metodi delle liste
print(f"Clienti sposati: {estados_civis.count('married')}")
````

**Esempio:** i valori letti da un file di testo sono sempre *stringhe*. Per fare calcoli, convertiamo ogni valore. Qui calcoliamo il saldo medio (`balance`) dei clienti:


````python
saldos_texto = arquivo_banco.extrair_coluna_por_nome(nome_coluna='balance')

# Convertiamo ogni saldo da str a int
saldos = []
for saldo in saldos_texto:
    saldos.append(int(saldo))

saldo_medio = sum(saldos) / len(saldos)
print(f'Saldo medio: {saldo_medio:.2f}')
````

> ⚠️ **Attenzione:** Python **non impedisce** di chiamare `arquivo_banco._extrair_conteudo()` dall'esterno della classe, ma ciò va contro la convenzione del `_`. I metodi interni possono cambiare o smettere di esistere in versioni future della classe; usa sempre gli attributi e i metodi pubblici.


> 💡 **Suggerimento:** questa classe è ottima per capire la programmazione orientata agli oggetti, ma nella pratica esistono strumenti già pronti per leggere CSV, come il modulo nativo [`csv`](https://docs.python.org/it/3/library/csv.html) e la libreria **Pandas**, che vedremo in moduli futuri. Entrambi sono costruiti proprio con classi e oggetti.

## 4. Ereditarietà

Ogni studente è una persona, ma non ogni persona è uno studente. L'**ereditarietà** permette alla classe `Estudante` di riutilizzare tutto ciò che la classe `Pessoa` già possiede, senza copiare codice, e di aggiungere solo ciò che è specifico di uno studente.


### 4.1 Definizione

L'ereditarietà crea una **specializzazione** di una classe. La classe originale è chiamata **classe madre** (o superclasse), e la nuova, **classe figlia** (o sottoclasse). La classe figlia eredita tutti gli attributi e i metodi della madre. La struttura generale è:

```python
class ClasseMae:

    def __init__(self, parametros):
        ...

class ClasseFilha(ClasseMae):

    def __init__(self, parametros):
        super().__init__(parametros_da_mae)
        self.atributo_3 = ...
        self.atributo_4 = ...

    def metodo_3(self, parametros):
        ...
```

| Elemento | Cosa fa |
| --- | --- |
| `class ClasseFilha(ClasseMae):` | Il nome della classe madre tra parentesi indica l'ereditarietà |
| `super()` | Dà accesso ai metodi della classe madre |
| `super().__init__(...)` | Esegue il costruttore della madre, che crea gli attributi ereditati |

> 💡 **Suggerimento:** un buon modo per capire se l'ereditarietà ha senso è la domanda **"è un"**: uno studente **è una** persona? Sì, quindi `Estudante` può ereditare da `Pessoa`.


### 4.2 Creare la classe `Estudante`

Riutilizziamo la classe `Pessoa` della sezione 2.5. Ogni studente studia in un'università, quindi prima creiamo la classe `Universidade`:


````python
class Universidade:

    def __init__(self, nome: str):
        self.nome = nome
````

Ora specializziamo la classe `Pessoa` nella classe `Estudante`. Il costruttore riceve i dati di una persona **più** l'università, passa i dati della persona al costruttore della madre con `super()` e crea l'unico nuovo attributo:


````python
class Estudante(Pessoa):

    def __init__(self, nome: str, idade: int, documento: str, universidade: Universidade):
        # La classe madre (Pessoa) crea gli attributi nome, idade e documento
        super().__init__(nome=nome, idade=idade, documento=documento)
        # Attributo esclusivo di Estudante
        self.universidade = universidade
````

> ⚠️ **Attenzione:** `super().__init__()` è obbligatorio quando la classe figlia ha un proprio `__init__`. Senza di esso, il costruttore della madre **non viene eseguito**, e gli attributi `nome`, `idade` e `documento` semplicemente non esistono nell'oggetto, generando un `AttributeError` la prima volta che vengono usati.


Nota un altro dettaglio: l'attributo `universidade` contiene un **oggetto** della classe `Universidade`. Questo è diverso dall'ereditarietà:

| Relazione | Domanda | Esempio | Come appare nel codice |
| --- | --- | --- | --- |
| Ereditarietà | "è un" | Uno studente **è una** persona | `class Estudante(Pessoa):` |
| Composizione | "ha un" | Uno studente **ha un'**università | `self.universidade = universidade` |


### 4.3 Manipolazione

Creiamo gli oggetti allo stesso modo di prima. Prima l'università, poi lo studente, che riceve l'oggetto università come parametro:


````python
usp = Universidade(nome='Universidade de São Paulo')
andre_estudante = Estudante(nome='Andre Perez', idade=30, documento='123', universidade=usp)
````

La classe `Estudante` non definisce `__str__`, né `falar()` né `dormir()`, ma l'oggetto riesce a usarli tutti, perché sono stati **ereditati** da `Pessoa`:


````python
# __str__ ereditato da Pessoa
print(andre_estudante)

# Metodi ereditati da Pessoa
andre_estudante.falar(texto='Ho un esame domani!')
andre_estudante.dormir(horas=2)
````

Per accedere a un attributo di un oggetto che si trova **dentro** un altro, concateniamo i punti:


````python
# andre_estudante.universidade è un oggetto Universidade; .nome è il suo attributo
print(andre_estudante.universidade.nome)
````

Uno studente è, allo stesso tempo, uno `Estudante` **e** una `Pessoa`. `isinstance()` lo conferma, e `issubclass()` verifica la relazione tra le classi:


````python
print(isinstance(andre_estudante, Estudante))
print(isinstance(andre_estudante, Pessoa))  # True: ogni Estudante è una Pessoa
print(isinstance(andre, Estudante))          # False: non ogni Pessoa è uno Estudante

print(issubclass(Estudante, Pessoa))
````

### 4.4 Nuovi metodi e sovrascrittura

Oltre a ereditare, la classe figlia può:

- **Aggiungere** metodi che hanno senso solo per lei (es: `estudar()`);
- **Sovrascrivere** (*override*) un metodo ereditato, definendo un'altra versione con lo stesso nome. Quando il metodo viene chiamato, Python usa la versione della classe figlia.

Miglioriamo `Estudante`: `__str__` ereditato non menziona l'università. Nella nuova versione, riutilizziamo il testo della madre con `super().__str__()` e aggiungiamo ciò che manca:


````python
class Estudante(Pessoa):

    def __init__(self, nome: str, idade: int, documento: str, universidade: Universidade):
        super().__init__(nome=nome, idade=idade, documento=documento)
        self.universidade = universidade

    def estudar(self, materia: str) -> None:
        # Metodo nuovo, esclusivo di Estudante
        print(f'{self.nome} sta studiando {materia} presso {self.universidade.nome}')

    def __str__(self) -> str:
        # Sovrascrittura: riutilizza il testo di Pessoa e aggiunge l'università
        return f'{super().__str__()}, studente presso {self.universidade.nome}'
````

Poiché la classe è stata ridefinita, ricreiamo l'oggetto (ricorda l'avviso della sezione 2.5):


````python
andre_estudante = Estudante(nome='Andre Perez', idade=30, documento='123', universidade=usp)

print(andre_estudante)  # usa __str__ di Estudante
print(andre)            # continua a usare __str__ di Pessoa

andre_estudante.estudar(materia='Programmazione Orientata agli Oggetti')
````

> 💡 **Suggerimento:** con l'ereditarietà, un miglioramento fatto nella classe madre arriva automaticamente a tutte le classi figlie. Se domani `Pessoa` guadagnasse un metodo `apresentar()`, ogni `Estudante` lo avrebbe anch'esso, senza nessuna riga in più.

## Riepilogo del Modulo

| Concetto | Cosa è | Sintassi |
| --- | --- | --- |
| Classe | Ricetta per creare oggetti | `class Pessoa:` |

| Sovrascrittura | Nuova versione di un metodo ereditato | ridefinire `__str__` nella classe figlia |
